# Exploração dos Dados

O conjunto de dados é composto por **12 variáveis**, das quais **11 são atributos físico-químicos do vinho** (variáveis independentes) e **1 é a variável alvo**, que representa a **qualidade sensorial do vinho**, avaliada por provadores especializados.

---

### Descrição das variáveis

1. **`fixed acidity` – Acidez fixa**  
   Refere-se à concentração de ácidos não voláteis no vinho, como o ácido tartárico e o ácido málico. Esses compostos contribuem para a estrutura, frescor e estabilidade do vinho.

2. **`volatile acidity` – Acidez volátil**  
   Mede a presença de ácidos voláteis, especialmente o ácido acético. Altas concentrações estão associadas a aromas desagradáveis, semelhantes ao vinagre.

3. **`citric acid` – Ácido cítrico**  
   Ácido orgânico naturalmente presente em pequenas quantidades. Intensifica o frescor e pode atuar como conservante natural.

4. **`residual sugar` – Açúcar residual**  
   Representa a quantidade de açúcar não fermentado que permanece no vinho após a fermentação. Afeta a doçura percebida da bebida.

5. **`chlorides` – Cloretos**  
   Refere-se principalmente à presença de cloreto de sódio e outros sais. Influencia o sabor salgado e a sensação na boca.

6. **`free sulfur dioxide` – Dióxido de enxofre livre**  
   Forma ativa do SO₂ que age como conservante e antioxidante. Protege o vinho de oxidação e contaminações microbianas.

7. **`total sulfur dioxide` – Dióxido de enxofre total**  
   Soma do SO₂ livre e combinado. Concentrações muito altas podem impactar negativamente o aroma e causar desconfortos em pessoas sensíveis.

8. **`density` – Densidade**  
   Relação entre massa e volume. Está relacionada ao teor alcoólico e ao açúcar residual, influenciando a viscosidade e a sensação na boca.

9. **`pH` – Potencial hidrogeniônico**  
   Mede a acidez ativa do vinho. Influencia diretamente a cor, sabor e estabilidade microbiológica da bebida.

10. **`sulphates` – Sulfatos**  
   Compostos que contribuem para a estabilidade microbiológica e podem afetar o sabor e a adstringência do vinho.

11. **`alcohol` – Teor alcoólico**  
   Indica a concentração de etanol resultante da fermentação. Está relacionado ao corpo, aroma e sensação térmica do vinho.

12. **`quality` – Qualidade**  
   Avaliação sensorial atribuída por especialistas, variando de 0 a 10, baseada em critérios como aroma, sabor, equilíbrio e corpo.

```python
import seaborn as sns
import matplotlib.pyplot as plt

def plot_histogram(df, column, title=None, nbins=50, histnorm='percent', marginal='box', opacity=0.8, color='#00CC96'):
    """
    Gera um histograma com boxplot marginal para uma coluna específica do DataFrame.

    Parâmetros:
    ----------
    df : pandas.DataFrame
        O DataFrame contendo os dados.
    column : str
        O nome da coluna a ser plotada no histograma.
    title : str, opcional
        O título principal do gráfico. Se None, será gerado automaticamente.
    nbins : int, opcional (default=50)
        Número de bins (intervalos) no histograma.
    histnorm : str, opcional (default='percent')
        Normalização do eixo y ('percent', 'probability', 'density', ou None).
    marginal : str, opcional (default='box')
        Gráfico adicional na lateral ('box', 'violin', ou None).
    opacity : float, opcional (default=0.8)
        Opacidade das barras do histograma.
    color : str, opcional (default='#00CC96')
        Cor das barras do histograma em formato HEX.

    Retorna:
    -------
    None
        Exibe o gráfico interativo do Plotly diretamente.
    """
    sns.set_theme(style="whitegrid")

    if title is None:
        title = f'Distribuição - {column.capitalize()}'

    # Figura principal
    fig, ax = plt.subplots(figsize=(10, 6))

    # Histograma
    sns.histplot(
        data=df,
        x=column,
        bins=nbins,
        kde=True,
        stat='percent' if histnorm == 'percent' else 'count',
        color=color,
        edgecolor='black',
        alpha=opacity,
        ax=ax
    )

    # Boxplot marginal (simulado com inset_axes)
    if marginal == 'box':
        box_ax = ax.inset_axes([0, 1.05, 1, 0.15], sharex=ax)
        sns.boxplot(
            data=df,
            x=column,
            ax=box_ax,
            color=color,
            width=0.3
        )
        box_ax.axis("off")

    elif marginal == 'violin':
        box_ax = ax.inset_axes([0, 1.05, 1, 0.2], sharex=ax)
        sns.violinplot(
            data=df,
            x=column,
            ax=box_ax,
            color=color,
            linewidth=0.5
        )
        box_ax.axis("off")

    # Ajustes
    ax.set_title(title, fontweight="bold", fontsize=16)
    ax.set_xlabel(column.capitalize(), fontsize=14)
    ax.set_ylabel("Porcentagem (%)" if histnorm == 'percent' else "Frequência", fontsize=14)
    ax.tick_params(axis='both', labelsize=12)

    plt.tight_layout()
    plt.show()
```

# Pré-processamento dos Dados

## Remoção do ID

A primeira providência no tratamento dos dados é retirar a coluna `Id`, uma vez que ela não configura influência no alvo.

```python
clean_dataset = dataset.copy()
clean_dataset = clean_dataset.drop(columns=['Id'])
clean_dataset.head()
```

## Winsorização

A Winsorização (ou Winsorization) é uma técnica de tratamento de outliers que consiste em limitar os valores extremos de uma variável numérica. Em vez de remover os outliers, você os "trava" em um limite definido.


```python
from scipy.stats.mstats import winsorize

def winsorize_iqr_scipy(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr
    lower_frac = (series < lower_limit).mean()
    upper_frac = (series > upper_limit).mean()

    return winsorize(series, limits=(lower_frac, upper_frac))

columns_with_outliers = [
    'fixed acidity',
    'volatile acidity',
    'citric acid',
    'residual sugar',
    'chlorides',
    'free sulfur dioxide',
    'total sulfur dioxide',
    'density',
    'sulphates',
    'alcohol'
]

winsorized_dataset = clean_dataset.copy()

for collumn in columns_with_outliers:
    winsorized_dataset[collumn] = winsorize_iqr_scipy(winsorized_dataset[collumn])
```

```python
winsorized_dataset.head()
```

## Padronização do Dados

A **padronização** é uma técnica de pré-processamento utilizada para transformar variáveis numéricas para uma escala comum, onde os dados passam a ter **média zero** e **desvio padrão igual a um**. Essa abordagem é especialmente útil em conjuntos de dados como o de vinhos, onde as variáveis possuem escalas distintas — por exemplo, o pH varia de 2 a 4, enquanto o dióxido de enxofre total pode ultrapassar 200. Padronizar os dados evita que atributos com maiores magnitudes dominem os algoritmos sensíveis à escala, como regressão linear, SVM, k-NN e PCA. Em Python, isso pode ser feito facilmente com a função `StandardScaler()` da biblioteca `sklearn.preprocessing`, que ajusta e transforma os dados de forma eficiente.


```python
from sklearn.preprocessing import StandardScaler

standardized_dataset = winsorized_dataset.copy()
num_cols = ['fixed acidity', 'volatile acidity', 'citric acid', 'residual sugar', 'chlorides', 'free sulfur dioxide', 'total sulfur dioxide', 'density', 'pH', 'sulphates', 'alcohol']

scaler = StandardScaler()
standardized_dataset[num_cols] = scaler.fit_transform(standardized_dataset[num_cols])
standardized_dataset.head()
```

## Agrupamento da Qualidade

Diante da distribuição altamente desbalanceada da variável `quality`, com poucas amostras nas pontas da escala (notas 3, 4 e 8), é recomendável realizar um **agrupamento dos níveis de qualidade** para facilitar a modelagem e melhorar a robustez dos resultados. Agrupar classes raras reduz o impacto do desbalanceamento, melhora a generalização dos modelos e evita que métricas como precisão e recall sejam dominadas por erros em classes com suporte muito pequeno. Uma estratégia comum e interpretável é consolidar as notas em três categorias: vinhos de **baixa qualidade** (notas 3 e 4), **qualidade média** (notas 5 e 6) e **alta qualidade** (notas 7 e 8). Essa abordagem transforma o problema em uma classificação com três classes mais equilibradas e com significado sensorial claro.

```python
grouped_dataset = standardized_dataset.copy()
grouped_dataset['quality'] = grouped_dataset['quality'].apply(lambda x: 0 if x <= 5 else (1 if x <= 6 else 2))
grouped_dataset.head()
```

# Treinamento dos Modelos

```python
from sklearn.model_selection import train_test_split

X = grouped_dataset.drop('quality', axis=1)
y = grouped_dataset['quality']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=37)
```

## KNN

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model = KNeighborsClassifier(
    n_neighbors=5,              # Número de vizinhos a considerar (K)
    weights='distance',         # 'uniform' ou 'distance' (pesa os vizinhos pela distância)
    algorithm='auto',           # Algoritmo: 'auto', 'ball_tree', 'kd_tree', 'brute'
    leaf_size=30,               # Tamanho da folha em algoritmos como kd_tree ou ball_tree
    p=2,                        # Potência da métrica de Minkowski (p=2 equivale à distância Euclidiana)
    metric='euclidean',         # Tipo de métrica: 'minkowski', 'euclidean', 'manhattan'
    metric_params=None,         # Parâmetros adicionais para a métrica (pode ser None)
    n_jobs=-1                   # Paralelização (-1 = todos os núcleos disponíveis)
)

knn_model.fit(X_train, y_train)
knn_predict = knn_model.predict(X_test)
```

## Árvore de Desição

```python
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(
    criterion='gini',               # Critério de divisão: 'gini' ou 'entropy'
    splitter='best',                # Estratégia de divisão: 'best' (ótima) ou 'random' (aleatória)
    max_depth=None,                 # Profundidade máxima da árvore (None = sem limite)
    min_samples_split=2,            # Mínimo de amostras para dividir um nó interno
    min_samples_leaf=1,             # Mínimo de amostras por folha
    min_weight_fraction_leaf=0.0,   # Fração mínima do peso total por folha (para dados com pesos)
    max_features=None,              # Máximo de features consideradas por divisão (None = todas)
    random_state=None,              # Semente para reprodutibilidade
    max_leaf_nodes=None,            # Número máximo de folhas (None = ilimitado)
    min_impurity_decrease=0.0,      # Redução mínima de impureza para realizar uma divisão
    class_weight=None,              # Peso das classes (None = igual para todas)
)

dt_model.fit(X_train, y_train)
dt_predict = dt_model.predict(X_test)
```

## Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,               # Número de árvores na floresta
    criterion='gini',               # Função de impureza: 'gini' ou 'entropy'
    max_depth=None,                 # Profundidade máxima da árvore (None = expande até todas as folhas)
    min_samples_split=2,            # Mínimo de amostras para dividir um nó
    min_samples_leaf=1,             # Mínimo de amostras em uma folha
    min_weight_fraction_leaf=0.0,   # Fração mínima de peso em uma folha
    max_features='sqrt',            # Número de features consideradas em cada split ('sqrt', 'log2', int ou float)
    max_leaf_nodes=None,            # Número máximo de folhas (None = ilimitado)
    min_impurity_decrease=0.0,      # Valor mínimo para uma divisão ser realizada
    bootstrap=True,                 # Se True, utiliza bootstrap para gerar amostras
    oob_score=False,                # Se True, calcula score OOB (out-of-bag)
    n_jobs=-1,                      # Paralelização (-1 = todos os núcleos disponíveis)
    random_state=37,                # Semente para reprodutibilidade
    verbose=0,                      # Se > 0, imprime logs durante o treinamento
    class_weight='balanced',        # Pode ser None ou 'balanced' para lidar com desbalanceamento
    warm_start=False                # Se True, reutiliza a solução anterior para adicionar mais árvores
)

rf_model.fit(X_train, y_train)
rf_predict = rf_model.predict(X_test)
```

## Stochastic Gradient Descent

```python
from sklearn.linear_model import SGDClassifier

sgd_model = SGDClassifier(
    loss='log_loss',            # Função de perda: 'hinge' (SVM), 'log_loss' (regressão logística), 'modified_huber', etc.
    penalty='l2',               # Regularização: 'l2', 'l1', ou 'elasticnet'
    alpha=0.0001,               # Força da regularização (quanto maior, mais penaliza os coeficientes)
    l1_ratio=0.15,              # Proporção l1/l2 se penalty='elasticnet'
    max_iter=1000,              # Número máximo de iterações sobre os dados
    tol=1e-3,                   # Tolerância para critério de parada
    shuffle=True,               # Embaralha os dados a cada época (recomendado)
    random_state=None,          # Semente para reprodutibilidade
    learning_rate='optimal',    # Esquema de taxa de aprendizado: 'constant', 'optimal', 'invscaling', 'adaptive'
    eta0=0.0,                   # Taxa de aprendizado inicial (usada quando learning_rate ≠ 'optimal')
    early_stopping=False,       # Para em validação se desempenho não melhorar (True = habilita)
    validation_fraction=0.1,    # Proporção da validação se early_stopping=True
    n_iter_no_change=5,         # Número de épocas sem melhoria para parar com early_stopping
    n_jobs=-1                   # Paralelização (se aplicável)
)

sgd_model.fit(X_train, y_train)
sgd_predict = sgd_model.predict(X_test)
```

## SVM

```python
from sklearn.svm import SVC

svc_model = SVC(
    C=0.8,                    # Parâmetro de penalização do erro: quanto maior, menor a margem e maior o risco de overfitting
    kernel='rbf',             # Tipo de kernel: 'linear', 'poly', 'rbf', 'sigmoid', ou função personalizada
    degree=3,                 # Grau do polinômio (usado somente se kernel='poly')
    gamma='scale',            # Coeficiente do kernel RBF/polynomial/sigmoid: 'scale', 'auto' ou valor numérico
    coef0=0.0,                # Término independente no kernel ('poly' e 'sigmoid')
    shrinking=True,           # Usa ou não a heurística de shrinking (True é padrão e mais eficiente)
    probability=False,        # Se True, habilita a estimativa de probabilidade (mais lento)
    tol=1e-3,                 # Tolerância para critério de parada
    cache_size=200,           # Tamanho do cache em MB
    class_weight=None,        # Pesos das classes (None = pesos iguais)
    verbose=False,            # Mostra ou não detalhes durante o treinamento
    max_iter=-1,              # Número máximo de iterações (-1 = ilimitado)
    random_state=None         # Semente de aleatoriedade (reprodutibilidade)
)

svc_model.fit(X_train, y_train)
svc_predict = svc_model.predict(X_test)
```

## Classificador AdaBoost

```python
from sklearn.ensemble import AdaBoostClassifier

adaboost_model = AdaBoostClassifier(
    estimator=None,           # Estimador base (None = usa DecisionTreeClassifier com max_depth=1)
    n_estimators=50,          # Número de estimadores fracos (iterações do boosting)
    learning_rate=1.0,        # Taxa de aprendizado (reduz o impacto de cada estimador fraco)
    random_state=None         # Semente para reprodutibilidade
)

adaboost_model.fit(X_train, y_train)
adaboost_predict = adaboost_model.predict(X_test)
```

## Classificador Gradient Boosting

```python
from sklearn.ensemble import GradientBoostingClassifier

gb_model = GradientBoostingClassifier(
    loss='log_loss',           # Função de perda: 'log_loss' (classificação), 'exponential' (como AdaBoost)
    learning_rate=0.1,         # Taxa de aprendizado (trade-off entre n_estimators e desempenho)
    n_estimators=100,          # Número de árvores (estimadores fracos) a serem adicionadas
    subsample=1.0,             # Fração da amostra usada para treinar cada árvore (subamostragem = ajuda a reduzir overfitting)
    criterion='friedman_mse',  # Critério de divisão das árvores: 'friedman_mse', 'squared_error', ou 'mae'
    min_samples_split=2,       # Mínimo de amostras para dividir um nó interno
    min_samples_leaf=1,        # Mínimo de amostras por folha
    max_depth=3,               # Profundidade máxima de cada árvore
    max_features=None,         # Número máximo de atributos usados em cada divisão
    random_state=None,         # Semente para reprodutibilidade
    max_leaf_nodes=None,       # Número máximo de folhas (None = ilimitado)
    warm_start=False,          # Se True, reusa a solução anterior para adicionar mais árvores (incremental)
    validation_fraction=0.1,   # Proporção para validação interna (usada com early_stopping)
    n_iter_no_change=None,     # Número de iterações sem melhoria para parar (usado com early_stopping)
    tol=1e-4,                  # Tolerância para parar (usado com early_stopping)
    verbose=0                  # Nível de verbosidade
)

gb_model.fit(X_train, y_train)
gb_predict = gb_model.predict(X_test)
```

# Avaliação dos Modelos

```python
from sklearn.metrics import classification_report, accuracy_score

predict_models = {
    'KNN': knn_predict,
    'Árvore de Decisão': dt_predict,
    'Random Forest': rf_predict,
    'SGD Classifier': sgd_predict,
    'SVM': svc_predict,
    'AdaBoost': adaboost_predict,
    'Gradient Boosting': gb_predict
}

for name, preds in predict_models.items():
    print("="*60)
    print(f'Classificador: {name}\n')
    print(f'\nAcurácia: {accuracy_score(y_test, preds):.4f}\n')
    print(classification_report(y_test, preds, digits=5))

print("="*60)
```

## Considerações sobre Resultados

Após a reclassificação da variável `quality` em três categorias balanceadas (baixa, média e alta), diversos algoritmos de classificação foram avaliados com base em métricas como acurácia, precisão, recall e f1-score. Os modelos que mais se destacaram em termos de desempenho geral foram o **KNN**, o **Random Forest** e o **Gradient Boosting**.

O **KNN** apresentou a maior acurácia (cerca de 70%), com f1-scores consistentes em todas as classes, especialmente nas classes majoritárias. No entanto, é um modelo sensível à dimensionalidade e menos eficiente com grandes volumes de dados, além de depender fortemente da distância entre amostras — o que pode ser afetado por ruídos ou features irrelevantes.

O **Gradient Boosting** também mostrou bom desempenho, com acurácia próxima de 67% e equilíbrio entre as classes. Entretanto, é mais propenso a overfitting e possui maior tempo de treinamento, o que pode dificultar ajustes rápidos em projetos iterativos.

O **Random Forest**, por sua vez, demonstrou **excelente desempenho geral**, com **acurácia de quase 70%**, f1-scores equilibrados e boa capacidade de prever corretamente as três classes. Além disso, é um modelo robusto contra overfitting, lida bem com variáveis não escaladas e fornece insights importantes como **importância das variáveis**.

Dado o bom desempenho, a estabilidade e a interpretabilidade, o **Random Forest** é o modelo mais indicado para ser refinado com técnicas de tuning de hiperparâmetros. O ajuste fino de parâmetros como número de árvores, profundidade máxima e número mínimo de amostras por folha pode levar a melhorias consideráveis no desempenho, especialmente para classes menos representadas.

## Ajuste e Otimização do Melhor Modelo

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint, uniform

rf_opt_model = RandomForestClassifier(
    n_jobs=-1,
    random_state=37,
    bootstrap=True,
    verbose=0
)

param_dist = {
    'n_estimators': randint(80, 350),
    'criterion': ['gini', 'entropy'],
    'min_samples_split': randint(2, 4),
    'min_samples_leaf': randint(1, 2),
    'max_features': ['sqrt', 'log2'],
    'min_impurity_decrease': uniform(0.0, 0.01),
    'oob_score': [True],
    'class_weight': ['balanced']
}

random_search = RandomizedSearchCV(
    estimator=rf_opt_model,
    param_distributions=param_dist,
    n_iter=50,
    scoring='f1_weighted',
    cv=11,
    verbose=1,
    n_jobs=-1,
    random_state=37
)

random_search.fit(X_train, y_train)
print(f'Melhores Parâmetros: {random_search.best_params_}')
```
