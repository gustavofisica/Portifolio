# Criação de Arrays no NumPy

O **array** é a estrutura de dados central do NumPy. Ele permite armazenar e manipular conjuntos de dados de maneira eficiente. Diferente das listas do Python, os arrays do NumPy são otimizados para operações matemáticas e consomem menos memória.


## Criando Arrays a Partir de Listas

Podemos criar um array NumPy a partir de uma lista do Python utilizando a função `np.array()`:

```python
lista = [i for i in range(0,10,1)]
print(lista)
```

```python
array = np.array(lista)
print(array)
```

```python
print(type(lista))
print(type(array))
```

## Atributos dos Arrays no NumPy

Os arrays do NumPy possuem diversos atributos que ajudam a entender suas propriedades. Entre os principais, destacam-se:

- **`.ndim`**: Retorna o número de dimensões do array.
- **`.shape`**: Retorna uma tupla indicando o tamanho de cada dimensão.
- **`.size`**: Retorna o número total de elementos no array.
- **`.dtype`**: Retorna o tipo de dado armazenado no array.
- **`.itemsize`**: Retorna o tamanho, em bytes, de cada elemento do array.
- **`.nbytes`**: Retorna o tamanho total, em bytes, ocupado pelo array na memória.

Abaixo, um exemplo para visualizar esses atributos na prática.

```python
print("Array:\n", array)
print("Número de dimensões (ndim):", array.ndim)
print("Formato (shape):", array.shape)
print("Número total de elementos (size):", array.size)
print("Tipo de dado (dtype):", array.dtype)
print("Tamanho de cada elemento em bytes (itemsize):", array.itemsize)
print("Tamanho total ocupado pelo array em bytes (nbytes):", array.nbytes)
```

## Arrays Multidimensionais

O NumPy permite a criação de arrays de múltiplas dimensões. Para isso, basta fornecer listas aninhadas:

```python
matriz = np.array([[lista], [lista]])
print(matriz)
```

## Criando Arrays com Valores Pré-definidos

O NumPy fornece funções para criar arrays com valores padrões:

* np.zeros(shape): Cria um array preenchido com zeros.
* np.ones(shape): Cria um array preenchido com uns.
* np.full(shape, valor): Cria um array preenchido com um valor específico.
* np.eye(n): Cria uma matriz identidade de tamanho n × n.

```python
zeros = np.zeros((3, 3))  # Matriz 3,3 de 0
print(zeros)
```

```python
ones = np.ones((2, 4))    # Matriz 2,4 de 1
print(ones)
```

```python
full = np.full((3, 3), 7)  # Matriz 3,3 de 7
print(full)
```

```python
identidade = np.eye(3)  # Matriz identidade 3x3
print(identidade)
```

## Criando Arrays com Intervalos Numéricos
O NumPy permite criar sequências numéricas de maneira eficiente:

* np.arange(início, fim, passo): Cria um array com valores espaçados uniformemente.
* np.linspace(início, fim, num): Cria um array com num elementos distribuídos igualmente entre início e fim com ponto flutuante.

```python
sequencia_inteiros = np.arange(0, 10, 2)
print(sequencia_inteiros)
```

```python
sequencia_flutuantes = np.linspace(0, 10, 20)
print(sequencia_flutuantes)
```

## Criando Arrays Aleatórios
Podemos gerar arrays com valores aleatórios utilizando o módulo * np.random:

* np.random.rand(dimensões): Gera números aleatórios entre 0 e 1.
* np.random.randn(dimensões): Gera números aleatórios com distribuição normal.
* np.random.randint(início, fim, tamanho): Gera números inteiros aleatórios em um intervalo.

```python
aleatorio = np.random.rand(3, 3)
print(aleatorio)
```

```python
aleatorio_normal = np.random.randn(3, 3)
print(aleatorio_normal)
```

```python
aleatorio_inteiros = np.random.randint(1, 10, (2, 2))
print(aleatorio_inteiros)
```

# Operações Básicas com Arrays no NumPy

O NumPy permite realizar operações matemáticas e lógicas diretamente sobre os arrays, sem a necessidade de laços `for`. Essas operações são vetorizadas, o que significa que são aplicadas a todos os elementos do array de forma eficiente.

## Operações Matemáticas

Os arrays do NumPy suportam operações matemáticas elementares, como:

- **Adição (`+`)**
- **Subtração (`-`)**
- **Multiplicação (`*`)** (elemento a elemento)
- **Divisão (`/`)**
- **Exponenciação** (** ou `np.power()`)
- **Módulo (`%`)**

```python
array_a = np.array([0, 1, 2, 3, 4])
array_b = np.random.randint(0, 10, 5)
print("Primeiro array")
print(array_a)
print("Segundo array")
print(array_b)
```

```python
print(array_a + array_b)
```

```python
print(array_a - array_b)
```

```python
print(array_a * array_b)
```

```python
print(array_a / array_b)
```

```python
print(array_a ** array_b)
```

```python
print(np.power(array_a, array_b))
```

```python
print(array_a % array_b)
```

## Funções Importantes no NumPy

Abaixo estão descritas diversas funções do NumPy utilizadas para análise, processamento e manipulação de arrays.

### 1. Funções Lógicas e de Verificação
- **`all`**: Retorna `True` se todos os elementos do array forem verdadeiros.
- **`any`**: Retorna `True` se pelo menos um elemento do array for verdadeiro.

### 2. Manipulação de Eixos
- **`apply_along_axis`**: Aplica uma função personalizada ao longo de um eixo específico de um array.

### 3. Funções Estatísticas e de Ordenação
- **`argmax`**: Retorna o índice do maior valor no array.
- **`argmin`**: Retorna o índice do menor valor no array.
- **`argsort`**: Retorna os índices que ordenariam o array.
- **`average`**: Calcula a média ponderada dos elementos do array.
- **`bincount`**: Conta a ocorrência de valores inteiros em um array.

### 4. Funções Matemáticas
- **`ceil`**: Arredonda os elementos para o inteiro superior mais próximo.
- **`clip`**: Restringe os valores de um array dentro de um intervalo específico.
- **`conj`**: Retorna o conjugado complexo dos elementos do array.
- **`corrcoef`**: Calcula o coeficiente de correlação entre variáveis.
- **`cov`**: Calcula a matriz de covariância.
- **`cross`**: Calcula o produto vetorial entre dois vetores.
- **`cumprod`**: Calcula o produto cumulativo dos elementos do array.
- **`cumsum`**: Calcula a soma cumulativa dos elementos do array.
- **`diff`**: Calcula a diferença discreta entre elementos consecutivos.
- **`dot`**: Calcula o produto escalar entre dois arrays.
- **`floor`**: Arredonda os elementos para o inteiro inferior mais próximo.
- **`inner`**: Calcula o produto interno entre arrays.
- **`invert`**: Inverte os bits de um array booleano ou inteiro.
- **`lexsort`**: Realiza uma ordenação indireta com base em colunas.

### 5. Estatísticas Básicas
- **`max`**: Retorna o maior valor do array.
- **`maximum`**: Compara dois arrays elemento a elemento e retorna o maior valor.
- **`mean`**: Calcula a média dos elementos do array.
- **`median`**: Retorna a mediana do array.
- **`min`**: Retorna o menor valor do array.
- **`minimum`**: Compara dois arrays elemento a elemento e retorna o menor valor.
- **`nonzero`**: Retorna os índices dos elementos não nulos.

### 6. Álgebra Linear e Transformações
- **`outer`**: Calcula o produto externo entre dois arrays.
- **`prod`**: Retorna o produto dos elementos do array.
- **`re`**: Retorna a parte real de números complexos em um array.
- **`round`**: Arredonda os elementos do array para o número especificado de casas decimais.
- **`sort`**: Ordena os elementos do array.
- **`std`**: Calcula o desvio padrão do array.
- **`sum`**: Retorna a soma dos elementos do array.
- **`trace`**: Retorna a soma dos elementos na diagonal principal de uma matriz.
- **`transpose`**: Transpõe as dimensões de um array.
- **`var`**: Calcula a variância dos elementos do array.
- **`vdot`**: Calcula o produto escalar entre dois arrays.
- **`vectorize`**: Converte uma função tradicional do Python em uma função vetorizada aplicável a arrays.
- **`where`**: Retorna os índices onde uma condição específica é atendida.


```python
arr = np.array([True, False, True])
print("all:", np.all(arr))  # False (Nem todos são True)
print("any:", np.any(arr))  # True (Pelo menos um é True)
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print("apply_along_axis (soma):", np.apply_along_axis(np.sum, axis=1, arr=arr))  # Soma por linha
```

```python
arr = np.array([3, 1, 4, 1, 5, 9])
print("argmax:", np.argmax(arr))  # Índice do maior valor
print("argmin:", np.argmin(arr))  # Índice do menor valor
print("argsort:", np.argsort(arr))  # Índices para ordenar o array
print("average:", np.average(arr, weights=[1, 2, 1, 2, 1, 1]))  # Média ponderada

arr = np.array([0, 1, 1, 2, 3, 1, 2])
print("bincount:", np.bincount(arr))  # Contagem de ocorrências dos valores inteiros
```

```python
arr = np.array([1.2, 3.7, -2.5, -4.1])
print("ceil:", np.ceil(arr))  # Arredonda para cima
print("floor:", np.floor(arr))  # Arredonda para baixo
print("clip:", np.clip(arr, -2, 3))  # Limita os valores entre -2 e 3

arr = np.array([1+2j, 3+4j])
print("conj:", np.conj(arr))  # Conjugado complexo

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
print("cross:", np.cross(arr1, arr2))  # Produto vetorial

arr = np.array([1, 2, 3, 4])
print("cumprod:", np.cumprod(arr))  # Produto cumulativo
print("cumsum:", np.cumsum(arr))  # Soma cumulativa

arr = np.array([10, 20, 30])
print("diff:", np.diff(arr))  # Diferença entre elementos consecutivos

arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])
print("dot:", np.dot(arr1, arr2))  # Produto escalar (matricial)
print("inner:", np.inner(arr1, arr2))  # Produto interno

arr = np.array([0, 1, 2, 3])
print("invert:", np.invert(arr))  # Inversão de bits

keys = np.array([2, 3, 1, 4])
values = np.array([40, 10, 30, 20])
print("lexsort:", np.lexsort((values, keys)))  # Ordenação por várias chaves
```

```python
arr = np.array([3, 1, 4, 1, 5, 9])
print("max:", np.max(arr))
print("maximum:", np.maximum(arr, 5))  # Elemento maior entre arr e 5
print("mean:", np.mean(arr))  # Média
print("median:", np.median(arr))  # Mediana
print("min:", np.min(arr))
print("minimum:", np.minimum(arr, 3))  # Elemento menor entre arr e 3

arr = np.array([[0, 1, 2], [3, 0, 4]])
print("nonzero:", np.nonzero(arr))  # Índices dos elementos não nulos
```

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
print("outer:", np.outer(arr1, arr2))  # Produto externo

arr = np.array([1, 2, 3, 4])
print("prod:", np.prod(arr))  # Produto dos elementos

arr = np.array([1+2j, 3+4j])
print("re:", np.real(arr))  # Parte real dos números complexos

arr = np.array([1.2, 2.5, 3.8])
print("round:", np.round(arr))  # Arredondamento padrão

arr = np.array([3, 1, 4, 1, 5, 9])
print("sort:", np.sort(arr))  # Ordenação

arr = np.array([1, 2, 3, 4])
print("std:", np.std(arr))  # Desvio padrão
print("sum:", np.sum(arr))  # Soma dos elementos

matriz = np.array([[1, 2], [3, 4]])
print("trace:", np.trace(matriz))  # Soma da diagonal principal

matriz = np.array([[1, 2], [3, 4]])
print("transpose:", np.transpose(matriz))  # Transposição da matriz

arr = np.array([1, 2, 3, 4])
print("var:", np.var(arr))  # Variância

arr1 = np.array([1, 2])
arr2 = np.array([3, 4])
print("vdot:", np.vdot(arr1, arr2))  # Produto escalar vetorial

def custom_function(x):
    return x ** 2 + 3

vectorized_func = np.vectorize(custom_function)
arr = np.array([1, 2, 3])
print("vectorize:", vectorized_func(arr))  # Aplica a função a cada elemento

arr = np.array([10, 20, 30, 40])
condicao = arr > 20
print("where:", np.where(condicao, "Maior que 20", "Menor ou igual a 20"))  # Retorna valores condicionais
```

## Indexação, Fatiamento e Iteração no NumPy

Manipular elementos de arrays no NumPy é essencial para processar dados de forma eficiente. Três operações fundamentais para acessar e modificar valores são **indexação (indexing)**, **fatiamento (slicing)** e **iteração (iterating).**

### 1. Indexação (Indexing)
A indexação permite acessar elementos individuais de um array. No NumPy, os índices começam em `0`, assim como nas listas do Python. Em arrays multidimensionais, utilizamos múltiplos índices para acessar elementos específicos.

### 2. Fatiamento (Slicing)
O fatiamento é usado para extrair subconjuntos de um array. Utiliza a sintaxe `[início:fim:passo]`, onde:
- `início` define o primeiro índice a ser incluído no fatiamento.
- `fim` define o índice onde o fatiamento termina (não incluso).
- `passo` indica de quantos em quantos elementos avançar.

Essa técnica permite selecionar partes de um array sem a necessidade de copiar manualmente os valores.

### 3. Iteração (Iterating)
A iteração ocorre quando percorremos os elementos de um array utilizando loops. No NumPy, o uso de **iterações vetorizadas** (sem `for` explícito) é preferível, pois torna as operações mais rápidas e eficientes. No entanto, quando necessário, podemos utilizar `for` para percorrer arrays unidimensionais e multidimensionais.

Com essas três operações, podemos manipular e acessar dados dentro dos arrays de forma flexível e otimizada.


```python
array = np.random.randint(0, 10, 15)
array
```

```python
array[2]
```

```python
array[1:4]
```

```python
array[:7:2]
```

```python
array[::-1]
```

```python
for i in array:
    print(i * 2)
```

## Manipulação de Forma (Shape Manipulation) no NumPy

A **forma** (shape) de um array no NumPy define suas dimensões, ou seja, o número de elementos em cada eixo. Manipular a forma de um array é útil para reorganizar os dados sem alterar seus valores.

### 1. Verificando a Forma de um Array
Podemos usar o atributo `.shape` para obter a forma de um array, que retorna uma tupla indicando o tamanho de cada dimensão.

### 2. Redimensionamento de Arrays (`reshape`)
A função `reshape()` permite alterar a forma do array sem modificar seus dados. O número total de elementos deve permanecer o mesmo.

### 3. Achatar um Array (`ravel` e `flatten`)
Se quisermos converter um array multidimensional em um array unidimensional, podemos usar:
- `ravel()`: Retorna uma visão do array original (modificações afetam o array original).
- `flatten()`: Retorna uma cópia do array original (modificações não afetam o original).

### 4. Transposição de Arrays (`transpose` e `.T`)
A transposição reorganiza os eixos do array, útil para operações de álgebra linear. Pode ser feita de duas formas:
- `transpose()`: Permite reorganizar os eixos manualmente.
- `.T`: Alterna automaticamente as dimensões.

### 5. Empilhamento de Arrays (Stacking)

O **stacking** combina arrays ao longo de um novo eixo. Algumas funções importantes incluem:

- `np.vstack()`: Empilha arrays verticalmente (linha a linha).
- `np.hstack()`: Empilha arrays horizontalmente (coluna a coluna).
- `np.dstack()`: Empilha arrays ao longo do terceiro eixo (profundidade).
- `np.concatenate()`: Combina arrays existentes ao longo de um eixo específico.

### 6. Divisão de Arrays (Splitting)

A divisão de arrays permite separar um array maior em vários menores. Algumas funções úteis são:

- `np.split()`: Divide um array em subarrays ao longo de um eixo.
- `np.hsplit()`: Divide horizontalmente (ao longo das colunas).
- `np.vsplit()`: Divide verticalmente (ao longo das linhas).
- `np.dsplit()`: Divide ao longo do terceiro eixo.

Com essas técnicas, é possível modificar a estrutura dos arrays para melhor adequação aos cálculos e operações numéricas.

```python
mask = (y == 3) | (y == 7)
X_subset = X_flattened[mask]
y_subset = y[mask]
```

```python
mean_3 = np.mean(X_subset[y_subset == 3], axis=0).reshape(28, 28)
mean_7 = np.mean(X_subset[y_subset == 7], axis=0).reshape(28, 28)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(mean_3, cmap="gray")
ax[0].set_title("Média do Dígito 3")
ax[1].imshow(mean_7, cmap="gray")
ax[1].set_title("Média do Dígito 7")
plt.show()
```

```python
import itertools

def distancia_euclidiana_media(X):
    num_amostras = X.shape[0]
    distancias = []

    for i, j in itertools.combinations(range(num_amostras), 2):
        dist = np.linalg.norm(X[i] - X[j])
        distancias.append(dist)

    return np.mean(distancias)
```

```python
X_3 = X_flattened[y == 3]
X_7 = X_flattened[y == 7]
```

```python
X_3_sample = X_3[:100]
X_7_sample = X_7[:100]
```

```python
distancia_3 = distancia_euclidiana_media(X_3_sample)
distancia_7 = distancia_euclidiana_media(X_7_sample)
```

```python
distancia_3_vs_7 = distancia_euclidiana_media(np.vstack((X_3_sample, X_7_sample)))
```

```python
print(f"Distância média entre imagens do 3: {distancia_3:.4f}")
print(f"Distância média entre imagens do 7: {distancia_7:.4f}")
print(f"Distância média entre imagens do 3 e do 7: {distancia_3_vs_7:.4f}")
```

## Interpretação dos Resultados

* Distância média entre imagens do 3 → 9.3611

    * Isso significa que, em média, a diferença entre duas imagens do número 3 é 9.36 unidades na escala Euclidiana.
    * Como os números escritos à mão variam um pouco entre diferentes pessoas, essa distância reflete essa variabilidade.

* Distância média entre imagens do 7 → 8.3575

    * As imagens do número 7 são ainda mais semelhantes entre si do que as do número 3.
    * A distância média menor (8.36) indica que os 7s têm menos variação no modo como são escritos.

* Distância média entre imagens do 3 e do 7 → 9.6197

    * Essa distância é maior do que as distâncias dentro das classes individuais (9.3611 para os 3s e 8.3575 para os 7s).
    * Isso confirma que, em média, um 3 é mais semelhante a outro 3 do que a um 7.
    * O mesmo vale para os 7s: eles são mais parecidos entre si do que com os 3s.
