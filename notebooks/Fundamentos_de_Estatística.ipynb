# Introdução à Estatísitca

Sabemos que computação é íntimamente ligada à matemática e talvez a área da matemática que tenha mais efeitos nas tecnologias computacionais seja a Estatística. Em suma, estatística é a ciência que coleta, organiza, analisa e interpreta dados, utilizando a teoria das probabilidades para modelar e prever eventos.

Além da própria computação, podemos estender o impacto por diversas outras áreas do conhecimento, como Física, Medicina, Engenharia, Economia e muitas outras, sendo fundamental para a tomada de decisões baseadas em dados.

Podemos dividi-la em três principais áreas:

* **Estatística Descritiva**: Foca na organização, resumo e apresentação de dados por meio de tabelas, gráficos e medidas numéricas, como média e desvio padrão.
* **Estatística Probabilística**: Utiliza modelos matemáticos baseados na teoria das probabilidades para descrever padrões e incertezas em fenômenos aleatórios.
* **Estatística Inferencial**: Permite tirar conclusões sobre uma população a partir de uma amostra, utilizando testes de hipóteses, intervalos de confiança e outros métodos analíticos.

Ao longo deste notebook, exploraremos esses conceitos de maneira prática, utilizando Python para manipulação e análise de dados.

## Dados

É difícil imaginar uma área da atuação humana que não gere dados que possibilitam análises para diversos objetivos. Esses dados, são base para o desenvolvimento da Estatística. Eles representam informações obtidas a partir de contagens, medidas, observações ou respostas. Quando organizados, os dados formam um conjunto de dados, que pode ser classificado em:

* População: Conjunto completo de todos os elementos de interesse em um estudo. Por exemplo, a altura de todos os habitantes de um país.
* Amostra: Subconjunto representativo da população, utilizado para realizar análises quando não é viável estudar todos os elementos. Por exemplo, a altura de 1.000 pessoas selecionadas aleatoriamente de uma cidade.

### Classificação das Variáveis
As variáveis são os atributos ou características dos elementos de um conjunto de dados. Elas podem ser classificadas em dois tipos principais:

1. **Variáveis Qualitativas (Categóricas)**: Representam características que não podem ser medidas numericamente, apenas classificadas em categorias.

    * **Nominais**: Não possuem uma ordem específica. Exemplo: cores de carros (vermelho, azul, preto).
    * **Ordinais**: Possuem uma ordem ou hierarquia. Exemplo: níveis de escolaridade (fundamental, médio, superior).

2. **Variáveis Quantitativas (Numéricas)**: Representam quantidades e podem ser medidas numericamente.

    * **Discretas**: Assumem valores inteiros e contáveis. Exemplo: número de filhos em uma família (0, 1, 2, 3…).
    * **Contínuas**: Podem assumir qualquer valor dentro de um intervalo, incluindo frações e decimais. Exemplo: altura de uma pessoa (1,75 m, 1,80 m).

A correta identificação e classificação das variáveis são fundamentais para escolher os métodos estatísticos adequados e interpretar corretamente os resultados das análises.

```python
import pandas as pd
import numpy as np
from IPython.display import display

# Gerando um conjunto de dados fictício para 100 alunos
np.random.seed(42)
num_alunos = 100

# Função para classificar o nível de aproveitamento com base na nota média
def classificar_aproveitamento(nota):
    if nota >= 9:
        return "Excelente"
    elif nota >= 8:
        return "Muito Bom"
    elif nota >= 7:
        return "Bom"
    elif nota >= 6:
        return "Regular"
    else:
        return "Insuficiente"

# Criando a tabela
dados_escolares = pd.DataFrame({
    "Nome": [f"Aluno_{i+1}" for i in range(num_alunos)],
    "Idade": np.clip(np.random.normal(12, 2, num_alunos), 6, 18).astype(int),
    "Faltas no Semestre": np.random.randint(0, 15, num_alunos),
    "Nota Média": np.clip(np.random.normal(7, 1, num_alunos), 5, 10)
})

# Aplicando a função de classificação
dados_escolares["Nível de Aproveitamento"] = dados_escolares["Nota Média"].apply(classificar_aproveitamento)

# Exibir a tabela
display(dados_escolares.head(10))  # Exibe apenas os 10 primeiros alunos
```

## Planejamento do Estudo Estatístico  

Para o sucesso de um estudo estatístico é essencial para garantir a precisão e a validade dos resultados. O processo pode ser dividido nas seguintes etapas:  

1. **Definição do Problema e Identificação da População e Variáveis**  
   - Determinar o objetivo do estudo e as perguntas que se deseja responder.  
   - Identificar a população-alvo e as variáveis que serão analisadas.  

2. **Planejamento da Coleta de Dados**  
   - Escolher a melhor metodologia para obter os dados (censos, amostras, experimentos, pesquisas, etc.).  
   - Definir o tamanho da amostra, caso a população completa não possa ser analisada.  
   - Considerar possíveis vieses que possam afetar a representatividade dos dados.  

3. **Coleta de Dados**  
   - Realizar a coleta conforme o plano estabelecido, garantindo que os dados sejam confiáveis e representativos.  
   - Armazenar e organizar os dados corretamente para facilitar a análise.  

4. **Descrição e Exploração dos Dados**  
   - Aplicar técnicas de estatística descritiva para resumir e visualizar os dados, como medidas de tendência central (média, mediana, moda) e dispersão (desvio padrão, variância).  
   - Utilizar tabelas, gráficos e histogramas para identificar padrões e tendências.  

5. **Análise e Tomada de Decisão**  
   - Aplicar métodos de estatística inferencial e probabilística para extrair conclusões e generalizar os resultados para a população.  
   - Utilizar testes de hipóteses, intervalos de confiança e modelos estatísticos para validar as conclusões.  

6. **Avaliação e Análise de Erros**  
   - Identificar possíveis fontes de erro na coleta e análise dos dados.  
   - Avaliar a margem de erro e a confiabilidade das conclusões.  
   - Ajustar o estudo, se necessário, para minimizar incertezas e aprimorar a precisão dos resultados.  

Seguir essas etapas garante que o estudo estatístico seja conduzido de maneira estruturada e confiável, aumentando a validade das conclusões obtidas.


## Amostragem  

Já vimos que a análise estatísticos são obtidos em uma população de dados. A partir dela, podemos retirar uma amostra que possa representar o todo da poulação. Esse processo de subdividir a população para uma amostra é chamado de **amostragem**. Seu objetivo é permitir a análise e extração de conclusões sobre a população sem a necessidade de examinar todos os seus elementos, o que pode ser inviável devido a limitações de tempo, custo ou acessibilidade.

Vamos tomar como exemplo, os dados educacionais da tabela que montamos anteriormente. Imagine que ela será usada para determinar a qualidade da educação num determinado país. Apesar da estrutura que a educação já disponibiliza, pode ser que usar todos os dados torne o trabalho árduo. Mas, se escolhermos aleatoreamente algumas escolas da rede a **estatística** permite que essa amostragem represente a população toda de estudantes.



### Tipos de Amostragem  

A amostragem pode ser classificada em dois grandes grupos:

1. Amostragem Probabilística  
Neste tipo de amostragem, todos os elementos da população têm uma chance conhecida e diferente de zero de serem selecionados. Isso garante maior representatividade e permite a generalização dos resultados. Os principais métodos são:

    * **Amostragem Aleatória Simples**: Cada elemento da população tem a mesma probabilidade de ser escolhido. Exemplo: sortear nomes de uma urna.
    Amostragem Aleatória Simples.png
    * **Amostragem Sistemática**: Seleciona-se um elemento inicial aleatório e, a partir dele, escolhem-se os próximos elementos seguindo um intervalo fixo. Exemplo: selecionar um cliente a cada 3 numa fila.
    Amostragem Aleatória Sistemática.png
    * **Amostragem Estratificada**: A população é dividida em grupos homogêneos (estratos) e uma amostra aleatória é retirada de cada grupo proporcionalmente ao seu tamanho. Exemplo: dividir uma escola por séries e selecionar aleatoriamente alunos de cada série.
    Amostragem Aleatória Estratificada.png
    * **Amostragem por Conglomerados**: A população é dividida em grupos (conglomerados) e apenas alguns desses grupos são escolhidos aleatoriamente para a análise. Exemplo: selecionar algumas escolas para representar todas as escolas de um estado.
    Amostragem por Conclomerados.png

2. Amostragem Não Probabilística  
Neste caso, a seleção dos elementos não é feita de maneira aleatória, podendo introduzir viés. Os principais métodos são:

- **Amostragem por Conveniência**: Seleciona os elementos mais fáceis de serem acessados. Exemplo: entrevistar apenas pessoas que passam em um shopping.  
- **Amostragem Intencional (ou por Julgamento)**: O pesquisador escolhe os elementos que considera mais representativos. Exemplo: selecionar especialistas para um estudo sobre inovação.  
- **Amostragem por Quotas**: Divide-se a população em grupos, mas a seleção dentro dos grupos não é aleatória, e sim controlada para atender proporções pré-definidas. Exemplo: garantir que uma pesquisa eleitoral tenha a mesma proporção de gênero e idade da população.  
- **Amostragem Bola de Neve**: Utilizada quando a população é difícil de acessar. Os primeiros participantes indicam novos participantes, formando uma rede. Exemplo: estudos com populações marginalizadas.  

É bom lembrar que existem muitos outros métodos de amostragem possíveis que não foram abordados nesse notebook.  

### Erros na Amostragem  
Mesmo em amostragens bem planejadas, podem ocorrer erros que afetam a precisão dos resultados. Os principais são:

- **Erro Amostral**: Diferença entre o resultado obtido na amostra e o valor real da população. Pode ser reduzido aumentando o tamanho da amostra.  
- **Viés de Seleção**: Ocorre quando a amostra não representa adequadamente a população. Exemplo: entrevistar apenas pessoas com acesso à internet em uma pesquisa sobre tecnologia.  
- **Viés de Não Resposta**: Ocorre quando algumas pessoas selecionadas não respondem, podendo distorcer os resultados.

# Distribuição de Frequência  

A distribuição de frequência é uma maneira de organizar e apresentar dados, permitindo visualizar quantas vezes cada valor ocorre dentro de um conjunto de dados. Ela pode ser representada por tabelas, gráficos ou histogramas.  

## Exemplo Prático  

Vamos considerar um conjunto de dados que montamos para representar os dados escolares.  

Podemos organizar esses dados em uma **tabela de frequência**, que contém:  

- **Frequência Absoluta $f_i$**: Número de vezes que cada valor aparece no conjunto de dados.  
- **Frequência Relativa $fr_i$**: Proporção da frequência absoluta em relação ao total de dados, expressa como porcentagem.  
- **Frequência Acumulada $F_i$**: Soma das frequências absolutas até determinado ponto.  


```python
# Criando a tabela de frequência
tabela_frequencia = dados_escolares["Nota Média"].value_counts().reset_index()
tabela_frequencia.columns = ["Nota", "Frequência Absoluta"]

# Ordenando as notas em ordem crescente
tabela_frequencia = tabela_frequencia.sort_values(by="Nota")

# Calculando a Frequência Relativa (%)
tabela_frequencia["Frequência Relativa (%)"] = (tabela_frequencia["Frequência Absoluta"] / len(dados_escolares)) * 100

# Calculando a Frequência Acumulada
tabela_frequencia["Frequência Acumulada"] = tabela_frequencia["Frequência Absoluta"].cumsum()

# Exibindo a tabela de forma organizada
display(tabela_frequencia)
```

## Largura da Classe  

Quando temos um conjunto grande de dados numéricos, uma forma eficiente de organizá-los é através da **distribuição de frequências em classes**. Nesse caso, os valores são agrupados em intervalos, chamados de **classes**, e a quantidade de dados dentro de cada classe é contabilizada.  

### Como calcular a Largura da Classe?  

A **largura da classe $h$** é o tamanho de cada intervalo de classe e pode ser calculada pela seguinte fórmula:

$$
h = \frac{\text{Valor Máximo} - \text{Valor Mínimo}}{\text{Número de Classes}}
$$

onde:  
- **Valor Máximo**: maior valor do conjunto de dados.  
- **Valor Mínimo**: menor valor do conjunto de dados.  
- **Número de Classes**: quantidade de intervalos que queremos dividir os dados (pode ser definido de acordo com a **Regra de Sturges**):

$$
k = 1 + 3.3 \log n
$$

onde $n$ é o número total de observações.

#### **Exemplo Prático**  

Vamos considerar as idades dos alunos do conjunto de dados que estamos analisando. A partir dessa coluna, calcularemos a largura de classe e construiremos a tabela de distribuição de frequências.


```python
import math

# Definir número de classes usando a Regra de Sturges
n = len(dados_escolares['Idade'])
k = round(1 + 3.3 * math.log10(n))  # Número de classes arredondado

# Calcular a largura da classe
valor_min = min(dados_escolares['Idade'])
valor_max = max(dados_escolares['Idade'])
largura_classe = math.ceil((valor_max - valor_min) / k)  # Arredondando para cima

# Criar os intervalos de classe
intervalos = [valor_min + i * largura_classe for i in range(k - 1)]

# Contar a frequência de cada classe
frequencias, bins = np.histogram(dados_escolares['Idade'], bins=intervalos)

# Criar a tabela de frequências
tabela_classes = pd.DataFrame({
    "Intervalo de Idade": [f"{int(bins[i])} - {int(bins[i+1]-1)}" for i in range(len(bins)-1)],
    "Frequência Absoluta": frequencias
})

# Adicionando frequência relativa (%)
tabela_classes["Frequência Relativa (%)"] = (tabela_classes["Frequência Absoluta"] / n) * 100

# Adicionando frequência acumulada
tabela_classes["Frequência Acumulada"] = tabela_classes["Frequência Absoluta"].cumsum()

# Exibindo os resultados
print(f"Número de classes: {k}")
print(f"Largura da classe: {largura_classe}")

# Exibir a tabela
display(tabela_classes)
```

## Ponto Médio  

O **ponto médio** de uma classe é o valor central do intervalo de classe. Ele é obtido pela média entre o limite inferior e o limite superior da classe:

$$
\text{Ponto Médio} = \frac{\text{Limite Inferior} + \text{Limite Superior}}{2}
$$

O ponto médio é uma medida importante na estatística descritiva, pois representa um valor típico dentro de cada classe e pode ser utilizado para cálculos estatísticos como média ponderada e medidas de tendência central.

### **Exemplo Prático**  

Vamos continuar com o exemplo das idades do grupo pessoas e calcular o ponto médio para cada intervalo de classe.


```python
# Calcular os pontos médios
pontos_medios = [(bins[i] + bins[i+1]) / 2 for i in range(len(bins)-1)]

# Criar a tabela de frequências
tabela_classes = pd.DataFrame({
    "Intervalo de Idade": [f"{int(bins[i])} - {int(bins[i+1]-1)}" for i in range(len(bins)-1)],
    "Frequência Absoluta": frequencias,
    "Ponto Médio": pontos_medios
})

# Adicionando frequência relativa (%)
tabela_classes["Frequência Relativa (%)"] = (tabela_classes["Frequência Absoluta"] / n) * 100

# Adicionando frequência acumulada
tabela_classes["Frequência Acumulada"] = tabela_classes["Frequência Absoluta"].cumsum()

# Exibir a tabela
display(tabela_classes)
```

## Histograma  

O **histograma** é um gráfico de barras utilizado para representar a distribuição de frequências de um conjunto de dados numéricos. Ele é muito útil para visualizar padrões, dispersão e a distribuição dos dados.

### **Características do Histograma**  
- **As barras são contíguas** (não há espaçamento entre elas), pois representam intervalos contínuos de valores.  
- **A altura das barras representa a frequência** (absoluta ou relativa) dos valores dentro de cada intervalo de classe.  
- **O eixo horizontal (X)** representa os intervalos de classe.  
- **O eixo vertical (Y)** representa a frequência dos valores em cada classe.  

### **Passos para construir um Histograma**  
1. Definir o número de classes (intervalos) com base na Regra de Sturges:  
   $$
   k = 1 + 3.3 \log n
   $$
   onde \( n \) é o número total de observações.  
2. Calcular a largura da classe:  
   $$
   h = \frac{\text{Valor Máximo} - \text{Valor Mínimo}}{\text{Número de Classes}}
   $$  
3. Criar uma tabela de frequência, agrupando os valores dentro dos intervalos de classe.  
4. Construir o histograma, onde cada barra representa um intervalo de classe e sua altura indica a frequência dos dados nesse intervalo.  

### **Exemplo Prático**  

Vamos utilizar o conjunto de dados de **idades** para construir um histograma.


```python
import matplotlib.pyplot as plt

# Criar o histograma
plt.figure(figsize=(8, 5))
plt.hist(dados_escolares['Idade'], bins=bins, edgecolor='black', alpha=0.7)

# Adicionar rótulos e título
plt.xlabel("Intervalos de Idade")
plt.ylabel("Frequência")
plt.title("Histograma das Idades")
plt.xticks(bins, rotation=45)  # Ajustar os rótulos do eixo X para mostrar as fronteiras
plt.grid(axis="y", linestyle="--", alpha=0.6)

# Exibir o histograma
plt.show()
```

## Polígono de Frequência  

O **polígono de frequência** é um gráfico que representa a distribuição de um conjunto de dados, conectando os pontos médios das classes por meio de segmentos de linha. Ele é útil para visualizar padrões e tendências dentro dos dados e pode ser utilizado para comparar diferentes distribuições.  

### **Características do Polígono de Frequência**  
- Baseia-se em uma **tabela de frequências**, onde os valores são agrupados em classes.  
- Utiliza os **pontos médios das classes** como referência para a posição dos pontos no eixo X.  
- A **frequência de cada classe** é representada no eixo Y.  
- Os pontos são conectados por uma linha para formar um gráfico contínuo.  
- Pode ser sobreposto a um histograma para facilitar a interpretação.  

### **Como Construir um Polígono de Frequência?**  
1. Criar uma tabela de frequências a partir do conjunto de dados.  
2. Determinar os **intervalos de classe** e calcular os **pontos médios** de cada classe.  
3. Calcular a **frequência** de cada intervalo.  
4. Plotar um gráfico de linhas conectando os pontos médios às suas respectivas frequências.  

### **Exemplo Prático**  

Agora, vamos construir um **polígono de frequência** para o conjunto de dados de idades.


```python
# Criar o polígono de frequência
plt.figure(figsize=(8, 5))
plt.plot(tabela_classes['Ponto Médio'], frequencias, marker='o', linestyle='-', color='b', label="Polígono de Frequência")

# Adicionar rótulos e título
plt.xlabel("Ponto Médio das Classes")
plt.ylabel("Frequência")
plt.title("Polígono de Frequência das Idades")
plt.xticks(tabela_classes['Ponto Médio'], rotation=45)  # Ajustar os rótulos do eixo X para mostrar os pontos médios
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.legend()

# Exibir o gráfico
plt.show()
```

## Ogiva  

A **ogiva** é um gráfico de frequência acumulada que representa a quantidade de observações abaixo de determinados valores em um conjunto de dados. Esse tipo de gráfico é útil para entender a distribuição cumulativa dos dados e identificar padrões como a mediana e percentis.  

### **Características da Ogiva**  
- Utiliza a **frequência acumulada** em vez da frequência simples.  
- O eixo **X** representa os **limites superiores** das classes.  
- O eixo **Y** representa a **frequência acumulada**.  
- Pode ser utilizada para identificar a distribuição cumulativa dos dados.  

### **Como Construir uma Ogiva?**  
1. Criar uma **tabela de frequências** com as classes do conjunto de dados.  
2. Calcular a **frequência acumulada** para cada classe.  
3. Usar os **limites superiores** das classes como referência no eixo **X**.  
4. Plotar um gráfico conectando os pontos para visualizar a distribuição cumulativa dos dados.  

### **Exemplo Prático**  

Agora, vamos construir uma **ogiva** para o conjunto de dados de idades.


```python
# Criar a ogiva
plt.figure(figsize=(8, 5))
plt.plot(bins[1:], tabela_classes['Frequência Acumulada'], marker='o', linestyle='-', color='r', label="Ogiva")

# Adicionar rótulos e título
plt.xlabel("Limite Superior das Classes")
plt.ylabel("Frequência Acumulada")
plt.title("Ogiva das Idades")
plt.xticks(bins, rotation=45)  # Ajustar os rótulos do eixo X para mostrar os limites superiores
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.legend()

# Exibir o gráfico
plt.show()
```

## Gráfico de Pareto  

O **Gráfico de Pareto** é uma ferramenta estatística usada para identificar os fatores mais importantes em um conjunto de dados. Ele permite visualizar quais categorias contribuem mais significativamente para um problema ou processo.  

Este gráfico segue o **Princípio de Pareto (Regra 80/20)**, que sugere que **aproximadamente 80% dos efeitos vêm de 20% das causas**. Ou seja, em muitas situações, algumas poucas categorias são responsáveis pela maior parte dos resultados.  

### **Características do Gráfico de Pareto**  
- **O eixo X** representa as categorias em ordem decrescente de frequência.  
- **O eixo Y esquerdo** mostra a **frequência absoluta** (quantidade de ocorrências de cada categoria).  
- **As barras** indicam a frequência individual de cada categoria.    

### **Quando usar o Gráfico de Pareto?**  
* Para identificar **principais causas de um problema** (por exemplo, defeitos em uma produção).  
* Para ajudar na **tomada de decisões** sobre onde focar esforços de melhoria.  
* Para entender **quais categorias têm mais impacto** sobre um determinado resultado.

# Medidas de Tendência Central  

As **Medidas de Tendência Central** são estatísticas que descrevem um valor típico dentro de um conjunto de dados, representando seu centro ou ponto de equilíbrio. Essas medidas são fundamentais na análise de dados, pois ajudam a resumir grandes quantidades de informações em um único valor representativo.  

As três principais medidas de tendência central são:  

- **Média**  
- **Mediana**  
- **Moda**   


## **Média (Média Aritmética)**  
A **média aritmética** é o valor obtido pela soma de todos os elementos do conjunto, dividida pela quantidade total de elementos.  

$$
\text{Média} = \frac{\sum X_i}{n}
$$

onde:  
- $X_i$ são os valores do conjunto de dados  
- $n$ é o número total de elementos  

**Exemplo:**  
Se temos as idades: **25, 30, 35, 40, 50**, a média será:  

$$
\frac{25 + 30 + 35 + 40 + 50}{5} = 36
$$


## **Mediana**  
A **mediana** é o valor central de um conjunto de dados ordenado. Caso o número de elementos seja **ímpar**, a mediana é o valor do meio. Se o número for **par**, a mediana será a média dos dois valores centrais.  

**Passos para calcular a mediana:**  
1. Ordenar os dados.  
2. Se houver um número ímpar de elementos, pegar o valor central.  
3. Se houver um número par de elementos, calcular a média dos dois valores centrais.  

**Exemplo:**  
Dados: **20, 25, 30, 35, 40, 50, 60**  
A mediana é **35**, pois é o valor central.  

Para dados pares: **20, 25, 30, 35, 40, 50**  
A mediana será a média de **30 e 35**:  

$$
\frac{30 + 35}{2} = 32.5
$$

## **Moda**  
A **moda** é o valor que mais se repete em um conjunto de dados.  

**Exemplo:**  
Se temos os números: **10, 20, 20, 30, 40, 40, 40, 50**, a moda é **40**, pois aparece com mais frequência.  

#### **Tipos de Moda**  
- **Unimodal**: Apenas um valor ocorre com mais frequência.  
- **Bimodal**: Dois valores aparecem com a mesma frequência máxima.  
- **Multimodal**: Mais de dois valores possuem a mesma frequência máxima.  
- **Amodal**: Quando nenhum valor se repete.  

#### **Exemplo Prático**  

Vamos usar o conjunto de idades para um exemplo prático.

```python
import statistics as stats
import seaborn as sns

# Cálculo das medidas de tendência central
media = np.mean(dados_escolares['Idade'])
mediana = np.median(dados_escolares['Idade'])
moda = stats.mode(dados_escolares['Idade'])

# Exibir os valores calculados
print(f"Média: {media:.2f}")
print(f"Mediana: {mediana}")
print(f"Moda: {moda}")
```

Podemos visualizar isso de forma gráfica

```python
# Criar um gráfico de distribuição para visualizar os dados
plt.figure(figsize=(8,5))
sns.histplot(dados_escolares['Idade'], bins=6, kde=True, color="skyblue", alpha=0.7, edgecolor="black")

# Adicionar linhas indicando Média, Mediana e Moda
plt.axvline(media, color='red', linestyle='--', label=f'Média ({media:.2f})')
plt.axvline(mediana, color='green', linestyle='-', label=f'Mediana ({mediana})')
plt.axvline(moda, color='blue', linestyle=':', label=f'Moda ({moda})')

# Configurar rótulos e título
plt.xlabel("Idades")
plt.ylabel("Frequência")
plt.title("Distribuição das Idades e Medidas de Tendência Central")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.6)

# Exibir o gráfico
plt.show()
```

## Forma das Distribuições  

A **forma da distribuição** de um conjunto de dados refere-se à maneira como os valores se organizam em torno da média. Isso pode ser analisado visualmente por meio de histogramas e estatisticamente com medidas como **assimetria (skewness)** e **curtose (kurtosis)**.  

As principais formas de distribuição são:


### **Distribuição Simétrica (Normal ou Gaussiana)**  
- A distribuição é **simétrica** em relação à média.  
- Média, mediana e moda têm valores **iguais ou muito próximos**.  
- O histograma tem um formato de **sino** característico da distribuição normal.   

```python
# Gerar dados de uma distribuição normal
dados_normal = np.random.normal(loc=50, scale=10, size=1000)

# Criar o gráfico
plt.figure(figsize=(8, 5))
sns.histplot(dados_normal, bins=30, kde=True, color="blue")
plt.title("Distribuição Simétrica")
plt.xlabel("Valores")
plt.ylabel("Frequência")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```

### **Distribuição Assimétrica Positiva (À Direita)**  
- A cauda da distribuição se estende para a **direita** (valores maiores).  
- A média é **maior** que a mediana, e a moda tem o menor valor.  
- Ocorre quando há **outliers altos**, que aumentam a média.    


```python
# Gerar dados de uma distribuição assimétrica positiva (exponencial)
dados_positiva = np.random.exponential(scale=10, size=1000) + 30

# Criar o gráfico
plt.figure(figsize=(8, 5))
sns.histplot(dados_positiva, bins=30, kde=True, color="red")
plt.title("Distribuição Assimétrica Positiva (À Direita)")
plt.xlabel("Valores")
plt.ylabel("Frequência")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```

### **Distribuição Assimétrica Negativa (À Esquerda)**  
- A cauda da distribuição se estende para a **esquerda** (valores menores).  
- A média é **menor** que a mediana, e a moda tem o maior valor.  
- Ocorre quando há **outliers baixos**, que reduzem a média.

```python
# Gerar dados de uma distribuição assimétrica negativa (inversão da exponencial)
dados_negativa = 100 - np.random.exponential(scale=10, size=1000)

# Criar o gráfico
plt.figure(figsize=(8, 5))
sns.histplot(dados_negativa, bins=30, kde=True, color="green")
plt.title("Distribuição Assimétrica Negativa (À Esquerda)")
plt.xlabel("Valores")
plt.ylabel("Frequência")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```

# Medidas de Dispersão  

As **Medidas de Dispersão** são estatísticas que indicam o quanto os valores de um conjunto de dados se afastam da média. Elas são fundamentais para entender a **variabilidade** dos dados e ajudam a complementar as **Medidas de Tendência Central** (média, mediana e moda).  

As principais medidas de dispersão são:


## **Amplitude**  
A **Amplitude** é a diferença entre o maior e o menor valor do conjunto de dados:  

$$
\text{Amplitude} = X_{\text{máx}} - X_{\text{mín}}
$$

- **Vantagem**: Fácil de calcular.  
- **Desvantagem**: Não considera como os valores estão distribuídos, sendo sensível a outliers.  

**Exemplo:**
Vamos identificar a amplitude de notas do nosso conjunto de dados escolares.

```python
# Calcular a amplitude
amplitude = max(dados_escolares['Nota Média']) - min(dados_escolares['Nota Média'])

# Exibir o resultado
print(f"Amplitude: {amplitude}")
```

## **Desvio Médio**  
O **Desvio Médio** mede o quão afastados os valores estão da média, utilizando o valor absoluto das diferenças:  

$$
\text{Desvio Médio} = \frac{\sum |X_i - \bar{X}|}{n}
$$  

onde:  
- $X_i$ são os valores do conjunto de dados  
- $\bar{X}$ é a média dos dados  
- $n$ é o número total de elementos  

**Exemplo:**

```python
# Calcular a média
media = np.mean(dados_escolares['Nota Média'])

# Calcular o desvio médio
desvio_medio = np.mean([abs(x - media) for x in dados_escolares['Nota Média']])

# Exibir o resultado
print(f"Média: {media:.2f}")
print(f"Desvio Médio: {desvio_medio:.2f}")
```

## **Variância**  
A **Variância** mede o quão dispersos os valores estão em relação à média. Ela é calculada elevando ao quadrado as diferenças entre os valores e a média:  

### **Para população:**
$$
\sigma^2 = \frac{\sum (X_i - \mu)^2}{N}
$$

### **Para amostra:**
$$
s^2 = \frac{\sum (X_i - \bar{X})^2}{n - 1}
$$  

onde:  
- $X_i$ são os valores do conjunto de dados  
- $\mu$ é a média populacional  
- $\bar{X}$ é a média amostral  
- $N$ é o tamanho da população  
- $n$ é o tamanho da amostra  

**Interpretação:**  
- Quanto **maior a variância**, maior a dispersão dos dados.  
- Quanto **menor a variância**, mais próximos os valores estão da média.  

**Exemplo:**

```python
# Calcular a variância da população
variancia_populacional = np.var(dados_escolares['Nota Média'])

# Calcular a variância da amostra (dividindo por n-1)
variancia_amostral = np.var(dados_escolares['Nota Média'], ddof=1)

# Exibir os resultados
print(f"Variância Populacional: {variancia_populacional:.2f}")
print(f"Variância Amostral: {variancia_amostral:.2f}")
```

## **Desvio Padrão**  

O **desvio padrão** é uma medida estatística usada para quantificar a dispersão ou variabilidade de um conjunto de dados em relação à sua média. Quanto maior o desvio padrão, maior é a variação dos dados em relação à média. Um desvio padrão pequeno indica que os dados estão próximos da média, enquanto um desvio padrão grande indica maior dispersão dos valores.

O desvio padrão pode ser calculado de duas maneiras:

- **Desvio padrão populacional** (quando se conhece toda a população):
$$
\sigma = \sqrt{\frac{\sum_{i=1}^{N}(X_i - \mu)^2}{N}}
$$

- **Desvio padrão amostral** (quando os dados são uma amostra da população, mais comum):
$$
s = \sqrt{\frac{\sum_{i=1}^{n}(X_i - \bar{X})^2}{n - 1}}
$$

onde:

- $ X_i $ é cada valor individual;
- $ \mu $ (populacional) ou $ \bar{X} $ (amostral) é a média;
- $ N $ é o tamanho da população;
- $ n $ é o tamanho da amostra.

**Exemplo:**

```python
# Calcular o desvio padrão populacional
desvio_padrao_populacional = np.std(dados_escolares['Nota Média'])

# Calcular o desvio padrão amostral (dividindo por n-1)
desvio_padrao_amostral = np.std(dados_escolares['Nota Média'], ddof=1)

# Exibir os resultados
print(f"Desvio Padrão Populacional: {desvio_padrao_populacional:.2f}")
print(f"Desvio Padrão Amostral: {desvio_padrao_amostral:.2f}")
```

## **Coeficiente de Variação (CV)**  

O **coeficiente de variação (CV)** é uma medida estatística que representa o desvio padrão em relação à média de um conjunto de dados, expresso em porcentagem. É usado principalmente para comparar a dispersão relativa entre diferentes conjuntos de dados, especialmente quando possuem unidades ou escalas diferentes.

O coeficiente de variação é calculado usando a seguinte fórmula:

$$
CV = \frac{\sigma}{\bar{X}} \times 100\%
$$  

- **Interpretação**:  
  - CV **baixo** indica que os valores estão próximos da média.  
  - CV **alto** indica que há grande variação nos dados.  

```python
# Calcular a média
media = np.mean(dados_escolares['Nota Média'])

# Calcular o desvio padrão
desvio_padrao = np.std(dados_escolares['Nota Média'], ddof=1)  # Amostral

# Calcular o Coeficiente de Variação (CV)
cv = (desvio_padrao / media) * 100

# Exibir o resultado
print(f"Média: {media:.2f}")
print(f"Desvio Padrão: {desvio_padrao:.2f}")
print(f"Coeficiente de Variação: {cv:.2f}%")
```

# Regra Empírica para Distribuições Simétricas

A **Regra Empírica**, também conhecida como **Regra 68-95-99.7**, descreve a distribuição dos dados em uma **distribuição normal** (simétrica). Ela afirma que:

- **68% dos dados** estão dentro de **1 desvio padrão** da média.
- **95% dos dados** estão dentro de **2 desvios padrões** da média.
- **99.7% dos dados** estão dentro de **3 desvios padrões** da média.

Essa regra é útil para prever **como os dados se comportam**, identificar **valores extremos (outliers)** e tomar decisões estatísticas.


### Como Funciona a Regra Empírica?

Se tivermos uma variável com distribuição normal, com média $\mu$ e desvio padrão $\sigma$, podemos afirmar:

- Aproximadamente **68% dos valores** estarão entre $\mu - \sigma$ e $\mu + \sigma$.
- Aproximadamente **95% dos valores** estarão entre $\mu - 2\sigma$ e $\mu + 2\sigma$.
- Aproximadamente **99.7% dos valores** estarão entre $\mu - 3\sigma$ e $\mu + 3\sigma$.

#### **Fórmulas da Regra Empírica**
$$
P(\mu - \sigma \leq X \leq \mu + \sigma) \approx 68\%
$$
$$
P(\mu - 2\sigma \leq X \leq \mu + 2\sigma) \approx 95\%
$$
$$
P(\mu - 3\sigma \leq X \leq \mu + 3\sigma) \approx 99.7\%
$$

O gráfico a seguir ilustra essa regra usando uma distribuição normal simulada.

```python
import scipy.stats as stats

# Definir a média e o desvio padrão
media = 50  # Média da distribuição
desvio_padrao = 10  # Desvio padrão

# Criar uma distribuição normal
x = np.linspace(media - 4 * desvio_padrao, media + 4 * desvio_padrao, 1000)
y = stats.norm.pdf(x, media, desvio_padrao)

# Criar o gráfico
plt.figure(figsize=(10, 6))
plt.plot(x, y, color="black", label="Distribuição Normal")

# Áreas da Regra Empírica
cores = ["blue", "green", "orange"]
limites = [1, 2, 3]
percentuais = [68, 95, 99.7]

for i in range(3):
    x_empirico = np.linspace(media - limites[i] * desvio_padrao, media + limites[i] * desvio_padrao, 100)
    y_empirico = stats.norm.pdf(x_empirico, media, desvio_padrao)
    plt.fill_between(x_empirico, y_empirico, color=cores[i], alpha=0.3,
                     label=f"{percentuais[i]}% dentro de ±{limites[i]}σ")

# Adicionar linhas de referência
plt.axvline(media, color="red", linestyle="--", label="Média")
plt.axvline(media - desvio_padrao, color="blue", linestyle="--")
plt.axvline(media + desvio_padrao, color="blue", linestyle="--")
plt.axvline(media - 2 * desvio_padrao, color="green", linestyle="--")
plt.axvline(media + 2 * desvio_padrao, color="green", linestyle="--")
plt.axvline(media - 3 * desvio_padrao, color="orange", linestyle="--")
plt.axvline(media + 3 * desvio_padrao, color="orange", linestyle="--")

# Configuração do gráfico
plt.title("Regra Empírica (68-95-99.7) para Distribuições Normais")
plt.xlabel("Valores")
plt.ylabel("Densidade de Probabilidade")
plt.legend()
plt.grid()

# Exibir gráfico
plt.show()
```

**A curva preta** representa uma **distribuição normal** centrada na média $\mu = 50$.  
**As áreas coloridas** representam os **intervalos da Regra Empírica**:  
   - 🔵 **Azul (68%)**: Representa os valores entre $\mu - \sigma$ e $\mu + \sigma$.
   - 🟢 **Verde (95%)**: Inclui a região entre $\mu - 2\sigma$ e $\mu + 2\sigma$.
   - 🟠 **Laranja (99.7%)**: Engloba os valores entre $\mu - 3\sigma$ e $\mu + 3\sigma$.

**Linhas tracejadas** indicam os limites dos desvios padrões.  

# Medidas de Posição  

As **Medidas de Posição** são estatísticas que indicam a localização relativa de um valor dentro de um conjunto de dados. Elas são utilizadas para dividir os dados em partes iguais e ajudar a entender a distribuição dos valores.  

As principais medidas de posição são:  
- **Quartis**  
- **Decis**  
- **Percentis**  

Essas medidas são amplamente utilizadas em análise de dados, relatórios estatísticos e avaliações de desempenho.


## Quartis  

Os **quartis** dividem um conjunto de dados ordenado em **quatro partes iguais**. Eles são representados por:  

- **1º quartil (Q1)**: 25% dos dados estão abaixo desse valor.  
- **2º quartil (Q2 ou Mediana)**: 50% dos dados estão abaixo desse valor.  
- **3º quartil (Q3)**: 75% dos dados estão abaixo desse valor.  

A **amplitude interquartil (AIQ)**, que mede a dispersão central dos dados, é calculada como:  

$$
AIQ = Q3 - Q1
$$

Os quartis são úteis para detectar valores extremos e entender a distribuição dos dados.

## Decis  

Os **decis** dividem o conjunto de dados ordenado em **dez partes iguais**. Eles seguem a mesma lógica dos quartis, mas com mais divisões:

- **D1**: 10% dos dados estão abaixo desse valor.  
- **D2**: 20% dos dados estão abaixo desse valor.  
- **D3**: 30% dos dados estão abaixo desse valor.  
- …  
- **D9**: 90% dos dados estão abaixo desse valor.  

Os decis são usados principalmente em análises econômicas e educacionais para comparar desempenhos relativos.

## Percentis  

Os **percentis** dividem o conjunto de dados ordenado em **cem partes iguais**. Eles são frequentemente usados em estatísticas de desempenho e avaliações, como testes escolares e estudos médicos.  

- **P10**: 10% dos dados estão abaixo desse valor.  
- **P25**: Equivalente ao **1º quartil (Q1)**.  
- **P50**: Equivalente à **mediana (Q2)**.  
- **P75**: Equivalente ao **3º quartil (Q3)**.  
- **P90**: 90% dos dados estão abaixo desse valor.  

Os percentis permitem analisar a posição de um valor dentro de uma distribuição e são muito usados em exames padronizados e análise de salários.

### Exemplo Prático

Agora, vamos calcular quartis, decis e percentis para o conjuto de salários da Empresa que usamos antes.

```python
# Calcular quartis
Q1 = np.percentile(dados_escolares['Nota Média'], 25)
Q2 = np.percentile(dados_escolares['Nota Média'], 50)  # Mediana
Q3 = np.percentile(dados_escolares['Nota Média'], 75)

# Calcular amplitude interquartil (AIQ)
AIQ = Q3 - Q1

# Calcular decis (D1, D5, D9)
D1 = np.percentile(dados_escolares['Nota Média'], 10)
D5 = np.percentile(dados_escolares['Nota Média'], 50)  # Igual à mediana
D9 = np.percentile(dados_escolares['Nota Média'], 90)

# Calcular percentis (P10, P50, P90)
P10 = np.percentile(dados_escolares['Nota Média'], 10)
P50 = np.percentile(dados_escolares['Nota Média'], 50)  # Igual à mediana
P90 = np.percentile(dados_escolares['Nota Média'], 90)

# Exibir resultados
print(f"1º Quartil (Q1): R$ {Q1:,.2f}")
print(f"Mediana (Q2): R$ {Q2:,.2f}")
print(f"3º Quartil (Q3): R$ {Q3:,.2f}")
print(f"Amplitude Interquartil (AIQ): R$ {AIQ:,.2f}\n")

print(f"1º Decil (D1): R$ {D1:,.2f}")
print(f"5º Decil (D5 - Mediana): R$ {D5:,.2f}")
print(f"9º Decil (D9): R$ {D9:,.2f}\n")

print(f"10º Percentil (P10): R$ {P10:,.2f}")
print(f"50º Percentil (P50 - Mediana): R$ {P50:,.2f}")
print(f"90º Percentil (P90): R$ {P90:,.2f}")
```

1. **Quartis**:
   - **Q1**: 25% dos salários estão abaixo desse valor.
   - **Q2 (Mediana)**: 50% dos salários estão abaixo desse valor.
   - **Q3**: 75% dos salários estão abaixo desse valor.
   - **AIQ**: Mede a dispersão entre Q1 e Q3.

2. **Decis**:
   - **D1**: 10% dos salários estão abaixo desse valor.
   - **D5 (Mediana)**: Metade dos salários estão abaixo desse valor.
   - **D9**: 90% dos salários estão abaixo desse valor.

3. **Percentis**:
   - **P10**: Indica um ponto onde apenas 10% dos salários são menores.
   - **P50 (Mediana)**: Representa o valor central da distribuição.
   - **P90**: Apenas 10% dos salários são maiores que esse valor.

## BoxPlot (Gráfico de Caixa)

O **BoxPlot**, ou **gráfico de caixa e bigodes**, é uma ferramenta estatística utilizada para visualizar a distribuição de um conjunto de dados. Ele mostra a dispersão, a mediana, os quartis e possíveis valores atípicos (*outliers*).

### Componentes de um BoxPlot

Um **BoxPlot** é composto por cinco elementos principais:

1. **Mínimo**: O menor valor dentro do limite normal da distribuição.
2. **1º Quartil (Q1 - 25%)**: O valor abaixo do qual se encontram 25% dos dados.
3. **Mediana (Q2 - 50%)**: O valor central dos dados.
4. **3º Quartil (Q3 - 75%)**: O valor abaixo do qual se encontram 75% dos dados.
5. **Máximo**: O maior valor dentro do limite normal da distribuição.

Além desses, um BoxPlot pode indicar:

- **Amplitude Interquartil (AIQ)**: Calculada como $Q3 - Q1$, representa a dispersão dos dados centrais.
- **Bigodes**: Extremos dos dados que não são considerados outliers.
- **Outliers**: Valores muito distantes da distribuição central, identificados como pontos isolados.


### Interpretação de um BoxPlot

O BoxPlot ajuda a:

* Identificar a **distribuição dos dados** (simétrica ou assimétrica).  
* Detectar **valores atípicos (outliers)**.  
* Comparar a dispersão de diferentes grupos de dados.  
* Analisar **tendências e variabilidade** dentro de um conjunto de valores.

Agora, vamos construir um **BoxPlot** para visualizar a distribuição de um conjunto de dados.

```python
# Criar BoxPlot
plt.figure(figsize=(8, 5))
sns.boxplot(x=dados_escolares['Nota Média'], color="skyblue")

# Adicionar rótulos e título
plt.title("BoxPlot - Distribuição Salarial")
plt.xlabel("Nota")
plt.grid(axis="x", linestyle="--", alpha=0.6)

# Exibir o gráfico
plt.show()
```

* **A caixa representa o intervalo entre o 1º quartil (Q1) e o 3º quartil (Q3)**.  
* **A linha dentro da caixa representa a mediana (Q2)**, o valor central dos dados.  
* **Os bigodes estendem-se até os valores máximo e mínimo dentro do limite normal da distribuição**.  
* **Os pontos fora dos bigodes são outliers**, indicando valores que diferem significativamente dos demais.  

Se a mediana estiver **centralizada** dentro da caixa, a distribuição é **simétrica**. Se estiver **deslocada para cima ou para baixo**, os dados são **assimétricos**.


# Análise de Outliers

Os **outliers** são valores que se destacam do restante dos dados, ou seja, estão significativamente acima ou abaixo da distribuição normal dos valores. Eles podem indicar erros, eventos raros ou padrões incomuns.


## Como identificar outliers?

Existem diferentes métodos para identificar outliers em um conjunto de dados. Os mais comuns são:

### 1. Regra do Intervalo Interquartil (IQR - Amplitude Interquartil)  
Considera um valor como outlier se ele estiver fora do intervalo:

$$
[Q1 - 1.5 \times AIQ, Q3 + 1.5 \times AIQ]
$$

Onde:  
- $Q1$ = 1º quartil (25%)  
- $Q3$ = 3º quartil (75%)  
- $AIQ = Q3 - Q1$ (Amplitude interquartil)  

### 2. Desvio Padrão  
Um valor é considerado outlier se estiver a mais de **2 ou 3 desvios padrões da média**:

$$
X < \bar{X} - 3\sigma \quad \text{ou} \quad X > \bar{X} + 3\sigma
$$

Onde:  
- $\bar{X}$ = Média dos dados  
- $\sigma$ = Desvio padrão  

### 3. BoxPlot  
O **gráfico de caixa** visualiza outliers como pontos individuais fora dos "bigodes" da distribuição.

## Como os outliers impactam a análise de dados?

* Podem indicar **erros de medição ou entrada** de dados.  
* Influenciam **média e desvio padrão**, tornando-os menos representativos.  
* Ajudam a identificar **fenômenos raros ou tendências incomuns**.  
* São essenciais na **detecção de fraudes e anomalias**.

# Introdução à Probabilidade  

A **probabilidade** é o ramo da matemática que mede a **chance de um evento ocorrer**. Ela é amplamente utilizada em estatística, ciência de dados, inteligência artificial, jogos, economia, finanças, e muitas outras áreas.  

A palavra **probabilidade** vem do latim _probabilitas_, que significa algo "provável" ou "plausível". A ideia central é medir **quantas vezes um evento pode ocorrer** em relação ao **número total de possibilidades**.


## Aplicações da Probabilidade  

* **Estatística**: Testes de hipóteses, previsão de tendências.  
* **Ciência de Dados**: Modelos preditivos e aprendizado de máquina.  
* **Jogos e Apostas**: Estratégias de roleta, poker, loteria.  
* **Finanças**: Modelagem de risco e previsão de investimentos.  
* **Medicina**: Probabilidade de diagnóstico e eficácia de tratamentos.  

Agora, vamos entender alguns conceitos fundamentais.

### **Experimento Aleatório**  
Um **experimento aleatório** é aquele cujo resultado não pode ser previsto com certeza antes de ser realizado.  
Exemplo:  
- Lançar um dado e obter um número.  
- Jogar uma moeda e ver se cai "cara" ou "coroa".  

### **Espaço Amostral ($S$)**  
O **espaço amostral** é o conjunto de **todos os resultados possíveis** de um experimento aleatório.  
Exemplo:  
- Para um dado: $S = \{1,2,3,4,5,6\}$.  
- Para uma moeda: $S = \{\text{cara}, \text{coroa}\}$.

### **Evento ($A$)**  
Um **evento** é um subconjunto do espaço amostral.  
Exemplo:  
- No lançamento de um dado, o evento **"sair um número par"** é $A = \{2,4,6\}$.  
- No lançamento de uma moeda, o evento **"sair cara"** é $A = \{\text{cara}\}$.  


### **Probabilidade ($P(A)$)**  
A probabilidade de um evento $A$ ocorrer é dada pela fórmula:

$$
P(A) = \frac{\text{número de casos favoráveis}}{\text{número total de casos possíveis}}
$$

A probabilidade varia entre **0 e 1**, onde:  
- **$P(A) = 0$** significa que o evento **nunca acontece**.  
- **$P(A) = 1$** significa que o evento **sempre acontece**.  
- **$0 < P(A) < 1$** significa que há uma certa chance do evento ocorrer.  

### Exemplo: Lançamento de um dado  
- O espaço amostral é $S = \{1,2,3,4,5,6\}$.  
- Se quisermos calcular a probabilidade de sair um número **par** ($A = \{2,4,6\}$):  

$$
P(A) = \frac{\text{número de casos favoráveis}}{\text{número total de casos possíveis}} = \frac{3}{6} = 0.5
$$

Isso significa que há **50%** de chance de sair um número par ao lançar um dado.

## Tipos de Probabilidade

A probabilidade pode ser classificada em diferentes categorias, dependendo da abordagem utilizada para calcular a chance de um evento ocorrer. Os principais tipos são:

1. **Probabilidade Clássica (Laplace)**
2. **Probabilidade Empírica (Frequentista)**
3. **Probabilidade Condicional**
4. **Probabilidade Geométrica**
5. **Probabilidade Bayesiana**
6. **Probabilidade Objetiva**
7. **Probabilidade Axiomática (Kolmogorov)**

Cada uma dessas abordagens tem aplicações específicas e ajuda a resolver diferentes tipos de problemas.


### Probabilidade Clássica (Laplace)

A **probabilidade clássica** se aplica quando **todos os eventos são igualmente prováveis**. Foi desenvolvida por **Pierre-Simon Laplace** e segue a seguinte fórmula:

$$
P(A) = \frac{\text{Número de casos favoráveis}}{\text{Número total de casos possíveis}}
$$

#### Exemplo: Lançamento de um dado

Se lançarmos um **dado de seis lados**, a chance de obter um número par é:

$$
P(\text{número par}) = \frac{3}{6} = 0.5
$$


### Probabilidade Empírica (Frequentista)

A **probabilidade empírica** é baseada na **frequência relativa** de um evento ao longo de múltiplas observações.

$$
P(A) = \frac{\text{Número de vezes que A ocorreu}}{\text{Número total de tentativas}}
$$

#### Exemplo: Lançamento de uma moeda 100 vezes

Se ao lançar uma moeda **100 vezes**, obtivermos **55 caras**, a probabilidade empírica será:

$$
P(\text{cara}) = \frac{55}{100} = 0.55
$$

Diferente da probabilidade clássica, essa abordagem se baseia em dados coletados.


### Probabilidade Condicional

A **probabilidade condicional** mede a chance de um evento ocorrer **dado que outro evento já aconteceu**.

$$
P(A | B) = \frac{P(A \cap B)}{P(B)}
$$

#### Exemplo: Sorteio de cartas

Se sabemos que a carta sorteada de um baralho é **vermelha**, qual é a probabilidade de ser um **ás**?

- No baralho, há **26 cartas vermelhas** e **2 ases vermelhos**.
  
$$
P(\text{ás | vermelho}) = \frac{2}{26} = 0.077
$$


### Probabilidade Geométrica

A **probabilidade geométrica** é usada quando os resultados possíveis são **medidos por comprimentos, áreas ou volumes**, e não apenas por contagens.

#### Exemplo: Escolher um ponto aleatório em um segmento

Se temos um segmento de **10 cm** e escolhemos um ponto ao acaso dentro dos **3 cm iniciais**, a probabilidade será:

$$
P(A) = \frac{\text{comprimento favorável}}{\text{comprimento total}} = \frac{3}{10} = 0.3
$$

Esse tipo de probabilidade é útil em problemas de **engenharia, física e geometria computacional**.


### Probabilidade Objetiva

A **probabilidade objetiva** é usada para descrever **eventos que ocorrem de forma determinística**, sendo derivada de leis físicas ou padrões observáveis.

#### Exemplo: Probabilidade de falha de um componente eletrônico

Se um fabricante sabe que **2% dos resistores produzidos falham**, então:

$$
P(\text{falha}) = 0.02
$$

Diferente da probabilidade subjetiva, essa abordagem é baseada em **dados estatísticos sólidos**.


### Probabilidade Axiomática (Kolmogorov)

A **probabilidade axiomática**, formulada por **Andrey Kolmogorov**, define a probabilidade em termos de três **axiomas matemáticos**:

1. A probabilidade de qualquer evento é um número **entre 0 e 1**.
   $$ 0 \leq P(A) \leq 1 $$
2. A probabilidade do **espaço amostral inteiro** é **1**.
   $$ P(S) = 1 $$
3. Se dois eventos são **mutuamente exclusivos**, então:
   $$ P(A \cup B) = P(A) + P(B) $$

Essa abordagem é a **base da teoria moderna da probabilidade** e usada para construir modelos matemáticos rigorosos.


# Teorema de Bayes

O **Teorema de Bayes** é um dos conceitos fundamentais da teoria das probabilidades e estatística. Ele permite atualizar a **probabilidade de um evento** com base em **informações adicionais**.

Esse teorema é amplamente utilizado em diversas áreas, como **inteligência artificial, aprendizado de máquina, diagnóstico médico, análise de risco, e detecção de fraudes**.


## Definição do Teorema de Bayes

O **Teorema de Bayes** estabelece que a probabilidade condicional de um evento $A$ ocorrer, dado que um evento $B$ já ocorreu, é dada por:

$$
P(A | B) = \frac{P(B | A) \cdot P(A)}{P(B)}
$$

onde:  
- $P(A | B)$ = **Probabilidade de A dado B** (Probabilidade posterior)  
- $P(B | A)$ = **Probabilidade de B dado A** (Verossimilhança)  
- $P(A)$ = **Probabilidade de A** (Probabilidade prévia)  
- $P(B)$ = **Probabilidade de B** (Evidência)

O **Teorema de Bayes** nos permite **atualizar a probabilidade de A** após observar que B ocorreu.


## Exemplo 1: Diagnóstico Médico

Imagine que um exame seja utilizado para detectar uma **doença rara** que afeta **0.5% da população**. O teste tem:
- **Sensibilidade de 95%** (probabilidade de um resultado positivo dado que a pessoa tem a doença).
- **Taxa de falso positivo de 5%** (probabilidade de um resultado positivo mesmo quando a pessoa não tem a doença).

Dado que um paciente recebeu um resultado **positivo**, qual a probabilidade de ele realmente estar doente?

### **Passo 1: Definir os eventos**
- $A$ = "O paciente tem a doença".
- $B$ = "O teste deu positivo".

### **Passo 2: Definir as probabilidades**
- $P(A) = 0.005$ (probabilidade de um indivíduo ter a doença).
- $P(B | A) = 0.95$ (sensibilidade do teste).
- $P(B | A^c) = 0.05$ (falso positivo, onde $A^c$ é "não tem a doença").
- $P(A^c) = 1 - P(A) = 0.995$ (probabilidade de não ter a doença).

### **Passo 3: Calcular $P(B)$ (Evidência)**
A probabilidade de um teste positivo é:

$$
P(B) = P(B | A) P(A) + P(B | A^c) P(A^c)
$$

$$
P(B) = (0.95 \times 0.005) + (0.05 \times 0.995)
$$

$$
P(B) = 0.00475 + 0.04975 = 0.0545
$$

### **Passo 4: Aplicar o Teorema de Bayes**
Agora, aplicamos a fórmula:

$$
P(A | B) = \frac{P(B | A) \cdot P(A)}{P(B)}
$$

$$
P(A | B) = \frac{0.95 \times 0.005}{0.0545} = \frac{0.00475}{0.0545} \approx 0.0871
$$

Mesmo com um **teste positivo**, a probabilidade de o paciente realmente ter a doença é **8.71%**. Isso ocorre porque a doença é **muito rara**, e a taxa de falsos positivos influencia a análise.


```python
# Função para calcular a Probabilidade usando o Teorema de Bayes
def teorema_de_bayes(P_A, P_B_dado_A, P_B_dado_Ac):
    P_Ac = 1 - P_A  # Complemento de A
    P_B = (P_B_dado_A * P_A) + (P_B_dado_Ac * P_Ac)  # Evidência
    P_A_dado_B = (P_B_dado_A * P_A) / P_B  # Aplicação do Teorema de Bayes
    return P_A_dado_B

# Exemplo do diagnóstico médico
P_A = 0.005  # Probabilidade da doença
P_B_dado_A = 0.95  # Sensibilidade do teste
P_B_dado_Ac = 0.05  # Taxa de falso positivo

probabilidade_final = teorema_de_bayes(P_A, P_B_dado_A, P_B_dado_Ac)
print(f"Probabilidade real de ter a doença dado um teste positivo: {probabilidade_final:.4f} ({probabilidade_final*100:.2f}%)")
```

## Exemplo 2: Detecção de Fraudes

Um banco utiliza um sistema para detectar **transações fraudulentas**. Sabemos que:
- **0.1% das transações são fraudes** ($P(F) = 0.001$).
- O sistema tem **98% de precisão para detectar fraudes** ($P(T | F) = 0.98$).
- O sistema tem **2% de falsos positivos** ($P(T | F^c) = 0.02$).

Se o sistema **detectou uma fraude ($T$)**, qual é a probabilidade de a transação realmente ser fraudulenta?

### **Passo 1: Definir os eventos**
- $F$ = "Transação é fraudulenta".
- $T$ = "Sistema detecta fraude".

### **Passo 2: Definir as probabilidades**
- $P(F) = 0.001$ (fraude real).
- $P(T | F) = 0.98$ (precisão do sistema).
- $P(T | F^c) = 0.02$ (falsos positivos).
- $P(F^c) = 1 - P(F) = 0.999$ (transação legítima).

### **Passo 3: Calcular $P(T)$ (Evidência)**
A probabilidade do sistema detectar fraude (corretamente ou erroneamente) é:

$$
P(T) = P(T | F) P(F) + P(T | F^c) P(F^c)
$$

$$
P(T) = (0.98 \times 0.001) + (0.02 \times 0.999)
$$

$$
P(T) = 0.00098 + 0.01998 = 0.02096
$$

### **Passo 4: Aplicar o Teorema de Bayes**
Agora, aplicamos a fórmula:

$$
P(F | T) = \frac{P(T | F) \cdot P(F)}{P(T)}
$$

$$
P(F | T) = \frac{0.98 \times 0.001}{0.02096} = \frac{0.00098}{0.02096} \approx 0.0468
$$

Mesmo que o sistema tenha identificado uma transação como **fraude**, a probabilidade real de que seja uma fraude é **4.68%**. Isso acontece porque as fraudes são muito raras e há falsos positivos.


```python
# Exemplo do detecção de fraudes
P_A = 0.001  # Fraude real
P_B_dado_A = 0.98  # Precisão do sistema
P_B_dado_Ac = 0.02  # Taxa de falso positivo

probabilidade_final = teorema_de_bayes(P_A, P_B_dado_A, P_B_dado_Ac)
print(f"Probabilidade real de seja uma fraude: {probabilidade_final:.4f} ({probabilidade_final*100:.2f}%)")
```

# Lei da Probabilidade Total

A **Lei da Probabilidade Total** é um princípio fundamental da estatística que permite calcular a **probabilidade de um evento**, considerando diferentes **cenários ou condições possíveis**.

Essa lei é útil quando um evento pode ocorrer de **várias formas distintas**, e queremos levar em conta todas as possibilidades.


## Definição da Lei da Probabilidade Total

Se um evento $B$ pode ocorrer sob **diferentes condições disjuntas** $A_1, A_2, ..., A_n$, então sua probabilidade é dada por:

$$
P(B) = P(B | A_1) P(A_1) + P(B | A_2) P(A_2) + ... + P(B | A_n) P(A_n)
$$

ou, de forma generalizada:

$$
P(B) = \sum_{i=1}^{n} P(B | A_i) P(A_i)
$$

onde:
- $A_1, A_2, ..., A_n$ são **eventos mutuamente exclusivos** que cobrem todo o espaço amostral.
- $P(A_i)$ é a **probabilidade de cada cenário ocorrer**.
- $P(B | A_i)$ é a **probabilidade de B ocorrer dentro do cenário A_i**.


### Exemplo: Teste de COVID-19

Um hospital realiza um **teste de COVID-19** em uma população dividida em **grupos de risco**:
- **10% das pessoas são do grupo de alto risco ($A_1$)**.
- **90% das pessoas são do grupo de baixo risco ($A_2$)**.

A precisão do teste varia conforme o grupo:
- Para **alto risco ($A_1$)**, a chance de um **teste positivo** ($B$) é **80%**.
- Para **baixo risco ($A_2$)**, a chance de um **teste positivo** ($B$) é **10%**.

A pergunta é: **Qual a probabilidade de um teste ser positivo, considerando toda a população?**

#### **Passo 1: Definir os eventos**
- $A_1$: Pessoa de alto risco (**10% da população**).
- $A_2$: Pessoa de baixo risco (**90% da população**).
- $B$: Teste positivo.

#### **Passo 2: Aplicar a Lei da Probabilidade Total**

$$
P(B) = P(B | A_1) P(A_1) + P(B | A_2) P(A_2)
$$

$$
P(B) = (0.8 \times 0.1) + (0.1 \times 0.9)
$$

$$
P(B) = 0.08 + 0.09 = 0.17
$$

A probabilidade de um **teste positivo** na população geral é **17%**.

```python
# Função para calcular a Probabilidade Total
def probabilidade_total(eventos, probabilidades, condicoes):
    P_B = sum([condicoes[i] * probabilidades[i] for i in range(len(eventos))])
    return P_B

# Exemplo 1: Teste de COVID-19
eventos = ["Alto Risco", "Baixo Risco"]
probabilidades = [0.1, 0.9]  # P(A1) e P(A2)
condicoes = [0.8, 0.1]  # P(B | A1) e P(B | A2)

resultado = probabilidade_total(eventos, probabilidades, condicoes)
print(f"Probabilidade de um teste positivo: {resultado:.2f} ({resultado*100:.2f}%)")

# Exemplo 2: Aprovação em um exame
eventos = ["Professor A", "Professor B"]
probabilidades = [0.4, 0.6]  # P(A1) e P(A2)
condicoes = [0.7, 0.8]  # P(B | A1) e P(B | A2)

resultado = probabilidade_total(eventos, probabilidades, condicoes)
print(f"Probabilidade de um aluno passar no exame: {resultado:.2f} ({resultado*100:.2f}%)")
```

# Distribuições de Probabilidade Discretas

As **distribuições de probabilidades discretas** descrevem a probabilidade associada a variáveis aleatórias que podem assumir um conjunto **finito** ou **enumerável** de valores.

Essas distribuições são amplamente utilizadas em estatística, inteligência artificial, ciência de dados e aprendizado de máquina para modelar eventos que ocorrem **em contagens inteiras**, como:
- **Número de chamadas em um call center por hora**.
- **Número de defeitos em uma linha de produção**.
- **Número de vezes que um time marca gols em um jogo**.

## Variável Aleatória Discreta

Uma **variável aleatória discreta** é aquela que pode assumir **valores inteiros específicos** (por exemplo, 0, 1, 2, ...). Sua probabilidade é descrita por uma **função de probabilidade**:

$$
P(X = x) = f(x)
$$

onde $X$ representa a variável aleatória e $f(x)$ é a função de probabilidade associada a cada valor de $X$.

Para uma distribuição discreta, a soma das probabilidades de todos os valores possíveis de $X$ deve ser **igual a 1**:

$$
\sum P(X = x) = 1
$$

## Principais Distribuições Discretas

As distribuições de probabilidade discreta mais comuns incluem:

1. **Distribuição de Bernoulli**
2. **Distribuição Binomial**
3. **Distribuição Poisson**
4. **Distribuição Geométrica**
5. **Distribuição Hipergeométrica**

Vamos explorar cada uma delas com exemplos.

### Distribuição de Bernoulli

A **distribuição de Bernoulli** modela experimentos com **duas possibilidades**: sucesso ($1$) ou fracasso ($0$), com probabilidade $p$ de sucesso e $1 - p$ de fracasso.

A função de probabilidade é:

$$
P(X = x) = p^x (1 - p)^{1 - x}, \quad x \in \{0,1\}
$$

#### Exemplo: Lançamento de uma moeda
- Se uma moeda tem **50% de chance de cair cara**, então:
  - $P(X=1) = 0.5$ (cara)
  - $P(X=0) = 0.5$ (coroa)


```python
from scipy.stats import bernoulli

# Parâmetro da distribuição (probabilidade de sucesso)
p = 0.5

# Gerar 1000 amostras da Distribuição de Bernoulli
data = bernoulli.rvs(p, size=1000)

# Plot do histograma
plt.figure(figsize=(8, 5))
sns.histplot(data, bins=2, discrete=True, kde=False)
plt.xticks([0, 1], ["Fracasso (0)", "Sucesso (1)"])
plt.xlabel("Resultado")
plt.ylabel("Frequência")
plt.title(f"Distribuição de Bernoulli (p={p})")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Distribuição Binomial

A **distribuição binomial** modela o número de sucessos em **n** tentativas independentes de um experimento de Bernoulli.

A função de probabilidade é:

$$
P(X = k) = \binom{n}{k} p^k (1 - p)^{n - k}
$$

onde:
- $n$ é o número de tentativas,
- $k$ é o número de sucessos desejado,
- $p$ é a probabilidade de sucesso em cada tentativa.

#### Exemplo: Lançamento de um dado
- Qual a probabilidade de obter **3 números 6** em **10 lançamentos** de um dado justo?
  - Aqui, $n = 10$, $k = 3$, e $p = \frac{1}{6}$.


```python
from scipy.stats import binom

# Parâmetros da distribuição
n = 10   # Número de tentativas
p = 1/6  # Probabilidade de sucesso em cada tentativa

# Gerar 1000 amostras da Distribuição Binomial
data = binom.rvs(n, p, size=1000)

# Plot do histograma
plt.figure(figsize=(8, 5))
sns.histplot(data, bins=n+1, discrete=True, kde=False)
plt.xlabel("Número de Sucessos")
plt.ylabel("Frequência")
plt.title(f"Distribuição Binomial (n={n}, p={p})")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Distribuição de Poisson

A **distribuição de Poisson** modela o número de eventos que ocorrem **em um intervalo de tempo fixo** quando os eventos ocorrem com uma **taxa média constante**.

A função de probabilidade é:

$$
P(X = k) = \frac{e^{-\lambda} \lambda^k}{k!}
$$

onde:
- $\lambda$ é a **média de eventos no intervalo**,
- $k$ é o número de eventos desejado.

#### Exemplo: Chamadas em um call center
- Se um call center recebe **média de 5 chamadas por minuto**, qual a probabilidade de receber exatamente **3 chamadas em um minuto**?


```python
from scipy.stats import poisson

# Parâmetro da distribuição (média de eventos por intervalo)
lamb = 5

# Gerar 1000 amostras da Distribuição de Poisson
data = poisson.rvs(lamb, size=1000)

# Plot do histograma
plt.figure(figsize=(8, 5))
sns.histplot(data, bins=15, discrete=True, kde=False)
plt.xlabel("Número de Eventos")
plt.ylabel("Frequência")
plt.title(f"Distribuição de Poisson (λ={lamb})")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Distribuição Geométrica

A **distribuição geométrica** modela o número de tentativas até o **primeiro sucesso** em um experimento de Bernoulli.

A função de probabilidade é:

$$
P(X = k) = (1 - p)^{k-1} p
$$

onde:
- $k$ é o número de tentativas até o primeiro sucesso,
- $p$ é a probabilidade de sucesso em cada tentativa.

#### Exemplo: Jogador acertar um gol
- Se um jogador tem **probabilidade de 30% de marcar um gol** em cada chute, qual a probabilidade de marcar **no terceiro chute**?


```python
from scipy.stats import geom

# Parâmetro da distribuição (probabilidade de sucesso)
p = 0.3

# Gerar 1000 amostras da Distribuição Geométrica
data = geom.rvs(p, size=1000)

# Plot do histograma
plt.figure(figsize=(8, 5))
sns.histplot(data, bins=15, discrete=True, kde=False)
plt.xlabel("Tentativas até o Primeiro Sucesso")
plt.ylabel("Frequência")
plt.title(f"Distribuição Geométrica (p={p})")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Distribuição Hipergeométrica

A **distribuição hipergeométrica** modela o número de sucessos **em uma amostra sem reposição**.

A função de probabilidade é:

$$
P(X = k) = \frac{\binom{K}{k} \binom{N-K}{n-k}}{\binom{N}{n}}
$$

onde:
- $N$ é o tamanho total da população,
- $K$ é o número total de elementos de interesse,
- $n$ é o tamanho da amostra,
- $k$ é o número de sucessos desejado.

#### Exemplo: Inspeção de qualidade
- Se há **100 peças**, sendo **20 defeituosas**, e **sorteamos 5 peças**, qual a probabilidade de encontrar **2 peças defeituosas**?


```python
from scipy.stats import hypergeom

# Parâmetros da distribuição
N = 100  # Tamanho da população total
K = 20   # Número de sucessos na população
n = 10   # Tamanho da amostra

# Gerar 1000 amostras da Distribuição Hipergeométrica
data = hypergeom.rvs(N, K, n, size=1000)

# Plot do histograma
plt.figure(figsize=(8, 5))
sns.histplot(data, bins=n+1, discrete=True, kde=False)
plt.xlabel("Número de Sucessos na Amostra")
plt.ylabel("Frequência")
plt.title(f"Distribuição Hipergeométrica (N={N}, K={K}, n={n})")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

# Distribuições de Probabilidade Contínuas

As **distribuições de probabilidade contínuas** descrevem variáveis aleatórias que podem assumir **qualquer valor dentro de um intervalo**. Diferente das distribuições discretas, que lidam com contagens inteiras, as distribuições contínuas modelam fenômenos como:

- **Altura e peso de pessoas**.
- **Tempo de espera em filas**.
- **Velocidade de carros em uma rodovia**.

## Variável Aleatória Contínua

Uma **variável aleatória contínua** pode assumir **infinitos valores** dentro de um intervalo real. Sua probabilidade não pode ser expressa por uma função de probabilidades discretas, mas sim por uma **função densidade de probabilidade (PDF - Probability Density Function)**.

A **probabilidade de uma variável contínua assumir um valor exato é sempre zero**, pois há infinitos valores possíveis. Em vez disso, trabalhamos com **probabilidades dentro de um intervalo**:

$$
P(a \leq X \leq b) = \int_{a}^{b} f(x) \,dx
$$

onde $f(x)$ é a **função densidade de probabilidade (PDF)**.

## Principais Distribuições Contínuas

As distribuições de probabilidade contínuas mais comuns incluem:

1. **Distribuição Uniforme**
2. **Distribuição Normal (Gaussiana)**
3. **Distribuição Exponencial**
4. **Distribuição Gama**
5. **Distribuição Beta**

Vamos explorar cada uma delas com exemplos.

### Distribuição Uniforme

A **distribuição uniforme** modela variáveis em que **todos os valores têm a mesma probabilidade** dentro de um intervalo $[a, b]$.

A função densidade de probabilidade (PDF) é:

$$
f(x) =
\begin{cases}
\frac{1}{b-a}, & \text{se } a \leq x \leq b \\
0, & \text{caso contrário}
\end{cases}
$$

#### Exemplo: Tempo de espera em um semáforo
- Se um semáforo abre entre **30 e 60 segundos aleatoriamente**, a probabilidade de um tempo qualquer dentro desse intervalo é **uniforme**.


```python
from scipy.stats import uniform

# Gerar dados da distribuição uniforme
a, b = 30, 60
data = uniform.rvs(a, b-a, size=10000)

# Plotar
plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, density=True, alpha=0.7)
plt.title("Distribuição Uniforme (30, 60)")
plt.xlabel("Valor")
plt.ylabel("Densidade")
plt.grid()
plt.show()
```

### Distribuição Normal (Gaussiana)

A **distribuição normal** é a mais importante da estatística e modela fenômenos naturais, como **altura de pessoas, notas de alunos e erros de medição**.

A PDF da distribuição normal é:

$$
f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{(x - \mu)^2}{2\sigma^2}}
$$

onde:
- $\mu$ = **média** da distribuição,
- $\sigma$ = **desvio padrão** (mede a dispersão dos valores).

A **curva normal** tem formato de **sino** e a **Regra Empírica (68-95-99.7%)** afirma que:
- **68% dos valores** estão dentro de $1\sigma$ da média.
- **95% dos valores** estão dentro de $2\sigma$ da média.
- **99.7% dos valores** estão dentro de $3\sigma$ da média.

#### Exemplo: Altura de pessoas
- Se a altura média de uma população é **170 cm** com **desvio padrão de 10 cm**, a distribuição das alturas segue uma curva normal.


```python
from scipy.stats import norm

# Gerar dados da distribuição normal
mu, sigma = 170, 10
data = norm.rvs(mu, sigma, size=10000)

# Plotar
plt.figure(figsize=(8, 5))
plt.hist(data, bins=50, density=True, alpha=0.7)
plt.title("Distribuição Normal (μ=170, σ=10)")
plt.xlabel("Valor")
plt.ylabel("Densidade")
plt.grid()
plt.show()
```

### Distribuição Exponencial

A **distribuição exponencial** modela o **tempo entre eventos sucessivos** que ocorrem a uma taxa constante, como:
- **Tempo até o próximo cliente chegar em um caixa**.
- **Tempo de vida de componentes eletrônicos**.

Sua função densidade de probabilidade é:

$$
f(x) = \lambda e^{-\lambda x}, \quad x \geq 0
$$

onde:
- $\lambda$ = **taxa de ocorrência** do evento.

#### Exemplo: Tempo até a próxima ligação em um call center
- Se um call center recebe em média **5 chamadas por minuto**, o tempo entre chamadas segue uma distribuição exponencial.


```python
from scipy.stats import expon

# Gerar dados da distribuição exponencial
lamb = 5
data = expon.rvs(scale=1/lamb, size=10000)

# Plotar
plt.figure(figsize=(8, 5))
plt.hist(data, bins=50, density=True, alpha=0.7)
plt.title("Distribuição Exponencial (λ=5)")
plt.xlabel("Tempo")
plt.ylabel("Densidade")
plt.grid()
plt.show()
```

### Distribuição Gama

A **distribuição gama** é uma generalização da **distribuição exponencial** e modela o tempo até a **ocorrência de k eventos**.

A PDF da distribuição gama é:

$$
f(x) = \frac{\lambda^k x^{k-1} e^{-\lambda x}}{(k-1)!}, \quad x \geq 0
$$

onde:
- $\lambda$ = taxa de ocorrência,
- $k$ = número de eventos desejados.

#### Exemplo: Tempo até a falha de um sistema
- Se um sistema tem uma taxa de falha de **1 falha a cada 10 horas**, a distribuição gama pode modelar o **tempo até a segunda falha**.


```python
from scipy.stats import gamma

# Parâmetros da distribuição
k = 2  # Número de eventos (forma)
theta = 2  # Escala (1/taxa)

# Gerar dados da Distribuição Gama
data = gamma.rvs(k, scale=theta, size=10000)

# Plotar
plt.figure(figsize=(8, 5))
plt.hist(data, bins=50, density=True, alpha=0.7)
plt.title(f"Distribuição Gama (k={k}, θ={theta})")
plt.xlabel("Tempo até k eventos")
plt.ylabel("Densidade")
plt.grid()
plt.show()
```

### Distribuição Beta

A **distribuição beta** é utilizada para modelar probabilidades e proporções, sendo amplamente usada em **estatística bayesiana**.

Sua função de densidade de probabilidade é:

$$
f(x) = \frac{x^{\alpha-1} (1-x)^{\beta-1}}{B(\alpha, \beta)}
$$

onde:
- $\alpha$ e $\beta$ são **parâmetros de forma**,
- $B(\alpha, \beta)$ é a **função beta normalizadora**.

#### Exemplo: Taxa de conversão de anúncios
- Se um site testa um novo design e coleta **20 cliques em 100 visitantes**, a distribuição beta modela a incerteza na **taxa real de conversão**.


```python
from scipy.stats import beta

# Parâmetros da distribuição
alpha = 2
beta_param = 5

# Gerar dados da Distribuição Beta
data = beta.rvs(alpha, beta_param, size=10000)

# Plotar
plt.figure(figsize=(8, 5))
plt.hist(data, bins=50, density=True, alpha=0.7)
plt.title(f"Distribuição Beta (α={alpha}, β={beta_param})")
plt.xlabel("Proporção")
plt.ylabel("Densidade")
plt.grid()
plt.show()
```

# Intervalo de Confiança

O **intervalo de confiança** é um conceito fundamental da estatística inferencial. Ele fornece um **intervalo de valores** onde, com certa **confiança**, esperamos que um parâmetro populacional verdadeiro (como a média ou a proporção) esteja contido.

Em vez de fornecer um único valor estimado para um parâmetro populacional, o intervalo de confiança oferece uma **margem de erro** que leva em conta a variação amostral.


## O que significa um Intervalo de Confiança?

Se um intervalo de confiança for **95%**, isso significa que, **se repetíssemos a amostragem muitas vezes**, 95% dos intervalos calculados conteriam o **valor real do parâmetro populacional**.

Exemplo:  
- Se dissermos que a altura média da população está entre **165 cm e 175 cm** com **95% de confiança**, significa que há **95% de chance** de que a verdadeira altura média da população esteja dentro desse intervalo.


## Como calcular um Intervalo de Confiança?

O intervalo de confiança para a **média populacional** é dado por:

$$
IC = \bar{x} \pm z \cdot \frac{\sigma}{\sqrt{n}}
$$

onde:
- $\bar{x}$ = **média amostral**,
- $z$ = **valor crítico da distribuição normal** para o nível de confiança desejado (ex.: 1.96 para 95%),
- $\sigma$ = **desvio padrão populacional** (se conhecido, caso contrário, usa-se o desvio amostral),
- $n$ = **tamanho da amostra**.

### Z-Score

O **Z-Score**, ou **escore padronizado**, mede **quantos desvios padrão um valor está da média** em uma **distribuição normal**.

O **Z-Score** é calculado como:

$$
Z = \frac{X - \mu}{\sigma}
$$

onde:
- **$X$** é o valor da variável,
- **$\mu$** é a média populacional,
- **$\sigma$** é o desvio padrão populacional.

O **Z-Score** nos diz **quantos desvios padrão um valor X está distante da média**. Ele é essencial para definir **intervalos de confiança** na **distribuição normal padrão ($\mu = 0, \sigma = 1$)**.

### Intervalo de Confiança e seus Níveis

O **nível de confiança** determina a **probabilidade de que o intervalo contenha o valor real do parâmetro populacional**. Os níveis mais comuns são:

| **Nível de Confiança** | **Z-Score** |
|----------------|------------|
| 80% | ±1.28 |
| 85% | ±1.44 |
| 90% | ±1.645 |
| 95% | ±1.96 |
| 99% | ±2.576 |
| 99.9% | ±3.291 |

Isso significa que, por exemplo:
- **95% dos valores** de uma distribuição normal estão dentro de **±1.96 desvios padrão** da média.
- **99% dos valores** estão dentro de **±2.576 desvios padrão**.

#### **Como interpretar o intervalo de confiança?**
Se um **intervalo de confiança for 95%**, isso significa que, **se repetíssemos a amostragem muitas vezes**, 95% dos intervalos conteriam o **valor real do parâmetro populacional**.

Exemplo:
- Se dissermos que a altura média da população está entre **165 cm e 175 cm** com **95% de confiança**, significa que há **95% de chance** de que a verdadeira altura média da população esteja dentro desse intervalo.

Agora, vamos visualizar **todos os níveis de confiança** em um gráfico!

```python
# Configuração dos níveis de confiança e seus valores críticos Z
niveis_confianca = [0.90, 0.95, 0.99]
cores = ['green', 'orange', 'red']

# Criando valores para a curva normal
x = np.linspace(-4, 4, 1000)
y = st.norm.pdf(x, 0, 1)  # Distribuição normal padrão (μ=0, σ=1)

# Criando o gráfico
plt.figure(figsize=(10, 6))
plt.plot(x, y, color='black', label='Distribuição Normal Padrão')

# Adicionando intervalos de confiança
for i, nivel in enumerate(niveis_confianca):
    z_score = st.norm.ppf((1 + nivel) / 2)
    x_conf = np.linspace(-z_score, z_score, 300)
    y_conf = st.norm.pdf(x_conf, 0, 1)

    plt.fill_between(x_conf, y_conf, color=cores[i], alpha=0.3, label=f"{nivel*100}% IC (Z={z_score:.2f})")
    plt.axvline(-z_score, color=cores[i], linestyle='dashed')
    plt.axvline(z_score, color=cores[i], linestyle='dashed')

# Ajustes no gráfico
plt.xlabel("Z-Score")
plt.ylabel("Densidade de Probabilidade")
plt.title("Distribuição Normal e Intervalos de Confiança para Diferentes Níveis")
plt.legend()
plt.grid()

# Exibir o gráfico
plt.show()
```

## Intervalo de Confiança para Diferentes Cenários

1. **Média com Desvio Padrão Populacional Conhecido**  
   - Utiliza a distribuição **normal** ($Z$-score).
   - Fórmula:  
     $$
     IC = \bar{x} \pm z \cdot \frac{\sigma}{\sqrt{n}}
     $$

2. **Média com Desvio Padrão Populacional Desconhecido**  
   - Utiliza a distribuição **t de Student**.
   - Fórmula:  
     $$
     IC = \bar{x} \pm t \cdot \frac{s}{\sqrt{n}}
     $$
   - Onde $s$ é o desvio padrão da amostra.

3. **Proporção Amostral**  
   - Para estimar proporções dentro de uma população.
   - Fórmula:  
     $$
     IC = \hat{p} \pm z \cdot \sqrt{\frac{\hat{p} (1 - \hat{p})}{n}}
     $$
   - Onde $\hat{p}$ é a **proporção amostral**.

## Exemplo 1: Intervalo de Confiança para a Média

Um pesquisador coleta uma amostra de **30 pessoas** e mede suas alturas. Ele encontra uma **média amostral de 170 cm** e um **desvio padrão de 10 cm**.

Pergunta: **Qual é o intervalo de confiança de 95% para a média populacional?**

### **Passo 1: Identificar os valores**
- $\bar{x} = 170$
- $\sigma = 10$
- $n = 30$
- Para um **IC de 95%**, o **valor crítico $z = 1.96$**.

### **Passo 2: Calcular o erro padrão**
$$
\frac{\sigma}{\sqrt{n}} = \frac{10}{\sqrt{30}} = \frac{10}{5.477} \approx 1.83
$$

### **Passo 3: Calcular o intervalo**
$$
IC = 170 \pm (1.96 \times 1.83)
$$

$$
IC = 170 \pm 3.58
$$

$$
IC = (166.42, 173.58)
$$

Podemos dizer, com **95% de confiança**, que a média da população está entre **166.42 cm e 173.58 cm**.


```python
import scipy.stats as st

# Dados
media_amostral = 170
desvio_padrao = 10
n = 30
nivel_confianca = 0.95

# Cálculo do intervalo de confiança
erro_padrao = desvio_padrao / np.sqrt(n)
z_score = st.norm.ppf((1 + nivel_confianca) / 2)  # Valor crítico para 95% (1.96)
margem_erro = z_score * erro_padrao

IC = (media_amostral - margem_erro, media_amostral + margem_erro)

print(f"Intervalo de Confiança de 95% para a média: {IC[0]:.2f} a {IC[1]:.2f}")
```

## Exemplo 2: Intervalo de Confiança para Proporções

Em uma pesquisa eleitoral, **200 pessoas** foram entrevistadas e **120 disseram que votariam em um candidato**. Qual é o **intervalo de confiança de 95%** para a proporção da população que votaria nesse candidato?

### **Passo 1: Identificar os valores**
- $\hat{p} = \frac{120}{200} = 0.6$ (60%).
- $n = 200$.
- Para **IC de 95%**, o **valor crítico $z = 1.96$**.

### **Passo 2: Calcular o erro padrão**
$$
\sqrt{\frac{\hat{p} (1 - \hat{p})}{n}} = \sqrt{\frac{0.6 \times 0.4}{200}}
$$

$$
\sqrt{\frac{0.24}{200}} = \sqrt{0.0012} \approx 0.0346
$$

### **Passo 3: Calcular o intervalo**
$$
IC = 0.6 \pm (1.96 \times 0.0346)
$$

$$
IC = 0.6 \pm 0.0678
$$

$$
IC = (0.532, 0.668)
$$

Com **95% de confiança**, a verdadeira proporção da população que votaria no candidato está entre **53.2% e 66.8%**.


```python
# Dados
n = 200
sucessos = 120
proporcao_amostral = sucessos / n
nivel_confianca = 0.95

# Cálculo do erro padrão
erro_padrao = np.sqrt((proporcao_amostral * (1 - proporcao_amostral)) / n)

# Valor crítico Z para 95%
z_score = st.norm.ppf((1 + nivel_confianca) / 2)
margem_erro = z_score * erro_padrao

IC = (proporcao_amostral - margem_erro, proporcao_amostral + margem_erro)

print(f"Intervalo de Confiança de 95% para a proporção: {IC[0]:.3f} a {IC[1]:.3f}")
```

## Tamanho da Amostragem

O **tamanho da amostragem** é um fator crítico na estatística inferencial, pois afeta diretamente a **precisão das estimativas** e a **confiabilidade dos resultados**. Quanto maior a amostra, **menor o erro amostral** e **mais representativa** ela tende a ser da população.

Esse parâmetro é importante porque:

1. **Maior amostra = maior precisão**  
   - Uma **amostra maior reduz o erro amostral**, tornando a estimativa mais confiável.
   
2. **Pequenas amostras podem levar a conclusões erradas**  
   - Se a amostra for muito pequena, pode haver **viés amostral**, o que leva a estimativas imprecisas.

3. **O tamanho adequado depende do contexto**  
   - Em **pesquisas eleitorais**, por exemplo, um tamanho de **1000 a 2000 entrevistas** costuma ser suficiente para estimar a intenção de voto em uma população grande.
   - Em **ensaios clínicos**, o tamanho da amostra deve ser calculado para garantir a **eficácia estatística** do estudo.

Para estimar uma **média populacional**, o tamanho da amostra é determinado pela fórmula:

$$
n = \left( \frac{Z \cdot \sigma}{E} \right)^2
$$

Onde:
- **$n$** = tamanho da amostra,
- **$Z$** = valor crítico da distribuição normal para o nível de confiança escolhido (ex.: 1.96 para 95%),
- **$\sigma$** = desvio padrão da população (se não for conhecido, pode-se usar uma estimativa amostral),
- **$E$** = erro máximo permitido (margem de erro).

Para estimar uma **proporção populacional**, usamos:

$$
n = \frac{Z^2 \cdot p \cdot (1 - p)}{E^2}
$$

Onde:
- **$p$** = proporção esperada da população (ex.: 50% = 0.5, se não soubermos a proporção exata).


### Exemplo 1: Tamanho da Amostra para uma Média

Um pesquisador quer estimar a **altura média da população**, com um **nível de confiança de 95%**, um **desvio padrão populacional de 10 cm**, e deseja que o erro máximo seja **2 cm**.

#### **Passo 1: Identificar os valores**
- **Z = 1.96** (para 95% de confiança),
- **$\sigma = 10$ cm**,
- **Erro máximo permitido ($E$) = 2 cm**.

#### **Passo 2: Aplicar a fórmula**
$$
n = \left( \frac{1.96 \times 10}{2} \right)^2
$$

$$
n = \left( \frac{19.6}{2} \right)^2
$$

$$
n = 9.8^2 = 96.04
$$

O pesquisador precisa de pelo menos **97 pessoas** na amostra para obter uma estimativa confiável.


```python
# Parâmetros do problema
z_score = st.norm.ppf(0.975)  # 95% de confiança (Z = 1.96)
desvio_padrao = 10  # Desvio padrão populacional
erro_permitido = 2  # Margem de erro

# Cálculo do tamanho da amostra
n = (z_score * desvio_padrao / erro_permitido) ** 2

# Exibir resultado
print(f"Tamanho da amostra necessário para estimar a média: {np.ceil(n):.0f}")
```

### Exemplo 2: Tamanho da Amostra para uma Proporção

Uma empresa quer estimar a **proporção de clientes satisfeitos** com um nível de confiança de **95%** e uma margem de erro de **5%**. Como **não sabemos a proporção real**, usamos **$p = 0.5$** (o caso mais conservador).

#### **Passo 1: Identificar os valores**
- **Z = 1.96** (para 95% de confiança),
- **$p = 0.5$** (máxima incerteza),
- **Erro máximo permitido ($E$) = 0.05**.

#### **Passo 2: Aplicar a fórmula**
$$
n = \frac{(1.96)^2 \times 0.5 \times (1 - 0.5)}{(0.05)^2}
$$

$$
n = \frac{3.8416 \times 0.25}{0.0025}
$$

$$
n = \frac{0.9604}{0.0025} = 384.16
$$

A empresa precisa entrevistar pelo menos **385 clientes** para obter um resultado confiável.


```python
# Parâmetros do problema
z_score = st.norm.ppf(0.975)  # 95% de confiança (Z = 1.96)
p = 0.5  # Proporção esperada (máxima incerteza)
erro_permitido = 0.05  # Margem de erro

# Cálculo do tamanho da amostra
n = (z_score ** 2 * p * (1 - p)) / (erro_permitido ** 2)

# Exibir resultado
print(f"Tamanho da amostra necessário para estimar a proporção: {np.ceil(n):.0f}")
```

# Testes de Hipóteses

Os **testes de hipóteses** são procedimentos estatísticos usados para **avaliar suposições sobre uma população** com base em uma **amostra de dados**. Eles ajudam a tomar **decisões estatísticas**, determinando se devemos **rejeitar ou não uma hipótese** com base nas evidências disponíveis.


## O que é uma Hipótese Estatística?

Uma **hipótese estatística** é uma **afirmação** sobre um parâmetro populacional, como a **média, proporção ou variância**.

Os testes de hipóteses envolvem duas hipóteses principais:

* **Hipótese Nula ($H_0$)**  
   - Representa o **status quo** (afirmação de que **não há efeito ou diferença**).  
   - Exemplo: *"A média dos salários dos engenheiros é R$ 5.000,00."*  

* **Hipótese Alternativa ($H_1$ ou $H_a$)**  
   - Representa uma **afirmação contrária** à hipótese nula.  
   - Exemplo: *"A média dos salários dos engenheiros é diferente de R$ 5.000,00."*  

#### Exemplo prático:
Queremos testar se um novo medicamento **reduz a pressão arterial**.

- **$H_0$**: O medicamento **não reduz** a pressão arterial.
- **$H_1$**: O medicamento **reduz** a pressão arterial.

## Passos de um Teste de Hipóteses

* **Definir as hipóteses** ($H_0$ e $H_1$).  
* **Escolher um nível de significância ($\alpha$)**.  
* **Coletar os dados e calcular a estatística de teste**.  
* **Determinar a região crítica ou o valor-p**.  
* **Tomar uma decisão** (rejeitar ou não $H_0$).

## Erros em Testes de Hipóteses

| Tipo de Erro | Descrição |
|-------------|------------|
| **Erro Tipo I (Falso Positivo)** | Rejeitar $H_0$ quando ela é verdadeira. |
| **Erro Tipo II (Falso Negativo)** | Não rejeitar $H_0$ quando ela é falsa. |

O **nível de significância** ($\alpha$) representa a **probabilidade de cometer um erro tipo I**.  
Geralmente usamos **$\alpha = 0.05$ (5%)**, o que significa que aceitamos um risco de 5% de **falsamente rejeitar $H_0$**.


## Tipos de Testes de Hipóteses

Os testes de hipóteses podem ser classificados em:

1. **Testes Paramétricos** (assumem que os dados seguem uma distribuição específica, como a normal)
   - **Teste Z** (para médias quando $\sigma$ é conhecido)
   - **Teste t de Student** (para médias com amostras pequenas e $\sigma$ desconhecido)
   - **Teste F** (para comparar variâncias)
   - **ANOVA** (para comparar mais de duas médias)

2. **Testes Não Paramétricos** (usados quando não podemos assumir distribuições específicas)
   - **Teste de Mann-Whitney** (equivalente ao teste t, mas sem exigir normalidade)
   - **Teste de Wilcoxon** (comparação de duas amostras dependentes)
   - **Teste de Qui-Quadrado** (para dados categóricos)

## Testes Unilaterais vs. Bilaterais

Os testes de hipóteses podem ser:

* **Teste Bilateral (Bicaudal)**  
   - Testa **se há qualquer diferença significativa** em ambas as direções.  
   - Exemplo: *"A nova máquina produz itens com um peso diferente do esperado?"*  
   - $H_0$: $\mu = 50$  
   - $H_1$: $\mu \neq 50$  

* **Teste Unilateral (Monocaudal)**  
   - Testa se há uma **diferença em apenas uma direção**.  
   - Exemplo: *"O novo medicamento reduz a pressão arterial?"*  
   - $H_0$: $\mu \geq 120$  
   - $H_1$: $\mu < 120$  

## Teste Z para uma Média

Usado quando o desvio padrão populacional ($\sigma$) é conhecido e n > 30.

```python
# Parâmetros
media_amostral = 48
media_populacional = 50
desvio_padrao_pop = 5
n = 40
alpha = 0.05  # Nível de significância

# Cálculo do Z-Score
z_score = (media_amostral - media_populacional) / (desvio_padrao_pop / np.sqrt(n))

# Valor crítico para teste bilateral (95% de confiança)
z_critico = st.norm.ppf(1 - alpha / 2)

# Decisão
if abs(z_score) > z_critico:
    print(f"Rejeitamos H0 (Z={z_score:.2f})")
else:
    print(f"Não rejeitamos H0 (Z={z_score:.2f})")
```

## Teste t para uma Média

Usado quando o desvio padrão populacional ($\sigma$) é desconhecido.

```python
# Dados da amostra
dados = [48, 52, 49, 50, 47, 53, 51, 49, 50, 52]  # Exemplo de amostra
media_populacional = 50
alpha = 0.05

# Cálculo do Teste t
t_score, p_valor = st.ttest_1samp(dados, media_populacional)

# Decisão
if p_valor < alpha:
    print(f"Rejeitamos H0 (p={p_valor:.3f})")
else:
    print(f"Não rejeitamos H0 (p={p_valor:.3f})")
```

## Teste Qui-Quadrado

Usado para dados categóricos.

```python
# Dados observados e esperados
observado = np.array([50, 30, 20])  # Exemplo de categorias
esperado = np.array([40, 40, 20])  # Distribuição esperada

# Cálculo do Qui-Quadrado
chi2, p_valor = st.chisquare(observado, esperado)

# Decisão
if p_valor < 0.05:
    print(f"Rejeitamos H0 (p={p_valor:.3f})")
else:
    print(f"Não rejeitamos H0 (p={p_valor:.3f})")
```

## Teste t para Duas Amostras Independentes

Usado para comparar médias de dois grupos diferentes.

```python
# Gerando dados fictícios para dois grupos
grupo_1 = np.random.normal(50, 5, 30)  # Média 50, desvio 5, 30 amostras
grupo_2 = np.random.normal(53, 5, 30)  # Média 53, desvio 5, 30 amostras

# Teste t para amostras independentes
t_score, p_valor = st.ttest_ind(grupo_1, grupo_2)

# Decisão
if p_valor < 0.05:
    print(f"Rejeitamos H0 (p={p_valor:.3f}) - Diferença significativa entre os grupos")
else:
    print(f"Não rejeitamos H0 (p={p_valor:.3f}) - Sem diferença significativa entre os grupos")
```

## Teste t para Amostras Pareadas

Usado quando os mesmos indivíduos são medidos antes e depois de um tratamento.

```python
# Criando um conjunto de dados antes e depois de um tratamento
antes = np.random.normal(50, 5, 20)
depois = antes + np.random.normal(-2, 2, 20)  # Supondo que o tratamento reduziu os valores

# Teste t para amostras pareadas
t_score, p_valor = st.ttest_rel(antes, depois)

# Decisão
if p_valor < 0.05:
    print(f"Rejeitamos H0 (p={p_valor:.3f}) - O tratamento teve efeito significativo")
else:
    print(f"Não rejeitamos H0 (p={p_valor:.3f}) - O tratamento não teve efeito significativo")
```

## Teste F para Comparação de Variâncias

Usado para comparar a variabilidade entre dois grupos.

```python
# Criando dois grupos com diferentes variâncias
np.random.seed(42)
grupo_1 = np.random.normal(50, 5, 30)  # Média 50, desvio 5
grupo_2 = np.random.normal(50, 10, 30)  # Média 50, desvio 10 (maior variância)

# Cálculo do Teste F
F = np.var(grupo_1, ddof=1) / np.var(grupo_2, ddof=1)
df1 = len(grupo_1) - 1
df2 = len(grupo_2) - 1
p_valor = st.f.cdf(F, df1, df2) if F < 1 else 1 - st.f.cdf(1/F, df1, df2)

# Exibir resultado
print(f"Estatística F: {F:.3f}")
print(f"p-valor: {p_valor:.3f}")

if p_valor < 0.05:
    print("Rejeitamos H0 - As variâncias são significativamente diferentes.")
else:
    print("Não rejeitamos H0 - As variâncias não são significativamente diferentes.")
```

## Teste ANOVA (Análise de Variância)

Usado para comparar mais de duas médias.

```python
# Criando três grupos com médias diferentes
np.random.seed(42)
grupo_1 = np.random.normal(50, 5, 30)
grupo_2 = np.random.normal(55, 5, 30)
grupo_3 = np.random.normal(60, 5, 30)

# Teste ANOVA
F, p_valor = st.f_oneway(grupo_1, grupo_2, grupo_3)

# Exibir resultado
print(f"Estatística F: {F:.3f}")
print(f"p-valor: {p_valor:.3f}")

if p_valor < 0.05:
    print("Rejeitamos H0 - Pelo menos um grupo é diferente.")
else:
    print("Não rejeitamos H0 - As médias são estatisticamente iguais.")
```

## Teste de Mann-Whitney (Teste Não Paramétrico para Duas Amostras)
Usado quando os dados não seguem distribuição normal.

```python
# Criando dois grupos sem normalidade
np.random.seed(42)
grupo_1 = np.random.exponential(50, 30)
grupo_2 = np.random.exponential(55, 30)

# Teste de Mann-Whitney
U, p_valor = st.mannwhitneyu(grupo_1, grupo_2, alternative='two-sided')

# Exibir resultado
print(f"Estatística U: {U:.3f}")
print(f"p-valor: {p_valor:.3f}")

if p_valor < 0.05:
    print("Rejeitamos H0 - Diferença significativa entre os grupos.")
else:
    print("Não rejeitamos H0 - Sem diferença significativa entre os grupos.")
```

## Teste de Wilcoxon (Teste Não Paramétrico para Amostras Dependentes)

Usado quando dados pareados não seguem distribuição normal.

```python
# Criando dados antes e depois sem normalidade
np.random.seed(42)
antes = np.random.exponential(50, 20)
depois = antes + np.random.exponential(2, 20)  # Pequena mudança

# Teste de Wilcoxon
T, p_valor = st.wilcoxon(antes, depois)

# Exibir resultado
print(f"Estatística T: {T:.3f}")
print(f"p-valor: {p_valor:.3f}")

if p_valor < 0.05:
    print("Rejeitamos H0 - O tratamento teve efeito significativo.")
else:
    print("Não rejeitamos H0 - O tratamento não teve efeito significativo.")
```

## Teste Bilateral

Usado para testar se há qualquer diferença (maior ou menor).

```python
# Criando um conjunto de dados
np.random.seed(42)
dados = np.random.normal(50, 5, 30)
media_populacional = 50
alpha = 0.05

# Teste t bilateral
t_score, p_valor = st.ttest_1samp(dados, media_populacional)

# Exibir resultado
print(f"Estatística t: {t_score:.3f}")
print(f"p-valor: {p_valor:.3f}")

if p_valor < alpha:
    print("Rejeitamos H0 - Existe diferença significativa.")
else:
    print("Não rejeitamos H0 - Sem diferença significativa.")
```

## Teste Unilateral (Esquerda ou Direita)

Usado para testar se um grupo é maior ou menor que um valor específico.

```python
# Criando um conjunto de dados
np.random.seed(42)
dados = np.random.normal(50, 5, 30)
media_populacional = 50
alpha = 0.05

# Teste t unilateral para valores menores (cauda esquerda)
t_score, p_valor = st.ttest_1samp(dados, media_populacional, alternative='less')

# Exibir resultado
print(f"Estatística t: {t_score:.3f}")
print(f"p-valor: {p_valor:.3f}")

if p_valor < alpha:
    print("Rejeitamos H0 - A média da amostra é significativamente menor.")
else:
    print("Não rejeitamos H0 - A média da amostra não é significativamente menor.")
```

# Erros em Modelos Estatísticos e Machine Learning

Os **erros** são fundamentais para a avaliação de modelos, pois indicam **o quão distante a previsão está da realidade**. Existem diferentes tipos de erros, dependendo do **tipo de problema** (classificação ou regressão).


## Tipos de Erros

Os principais tipos de erros em **modelos estatísticos e de machine learning** são:

### **Erros em Classificação**
Em problemas de **classificação**, os erros podem ser analisados a partir da **matriz de confusão**, que classifica as previsões como:

| **Predição vs Realidade** | **Positivo (Real)** | **Negativo (Real)** |
|---------------------------|-------------------|-------------------|
| **Positivo (Previsto)**   | Verdadeiro Positivo (**TP**) | Falso Positivo (**FP**) |
| **Negativo (Previsto)**   | Falso Negativo (**FN**) | Verdadeiro Negativo (**TN**) |

Os erros em classificação incluem:

✔ **Falso Positivo (FP)** → O modelo previu **positivo**, mas era **negativo**.  
✔ **Falso Negativo (FN)** → O modelo previu **negativo**, mas era **positivo**.  

**Exemplo prático**:  
- Se um teste de **COVID-19** dá **positivo para uma pessoa saudável**, isso é um **falso positivo**.  
- Se um teste de **câncer** dá **negativo para um paciente com a doença**, isso é um **falso negativo**.

#### **Erros Tipo I e Tipo II**
Os falsos positivos e falsos negativos também são conhecidos como:
- **Erro Tipo I (Falso Positivo)**: Rejeitar a hipótese nula quando ela é verdadeira.  
- **Erro Tipo II (Falso Negativo)**: Não rejeitar a hipótese nula quando ela é falsa.

Quanto **menor o Erro Tipo I, maior o Erro Tipo II**, e vice-versa. Por isso, escolher um **nível de significância adequado** é fundamental.


### **Erros em Regressão**
Em problemas de **regressão**, os erros são medidos como a **diferença entre o valor previsto ($\hat{y}$) e o valor real ($y$)**.

Os principais erros são:

✔ **Erro Residual** → Diferença entre o valor real e a previsão.  

$$
Erro = y - \hat{y}
$$

✔ **Erro Absoluto Médio (MAE)** → Média dos erros absolutos.  

$$
MAE = \frac{1}{n} \sum_{i=1}^{n} | y_i - \hat{y}_i |
$$

✔ **Erro Quadrático Médio (MSE)** → Média dos erros elevados ao quadrado.  

$$
MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

✔ **Erro Quadrático Médio Root (RMSE)** → Raiz quadrada do MSE.  

$$
RMSE = \sqrt{MSE}
$$

**Exemplo prático**:  
- Se um modelo de previsão do tempo prevê **25°C** e a temperatura real foi **30°C**, o erro é **5°C**.  
- Se o erro médio do modelo for **3°C**, isso significa que, em média, a previsão desvia **3 graus** da realidade.


### **Viés e Variância**
O erro total de um modelo pode ser decomposto em **viés e variância**.

#### **Viés**
- Mede o **quanto as previsões estão sistematicamente erradas**.  
- Um modelo com **alto viés** tende a **subajustar** (underfitting) os dados.  

* **Exemplo**: Um modelo muito simples que sempre prevê **25°C** para qualquer dia.

#### **Variância**
- Mede **o quanto as previsões variam em diferentes amostras**.  
- Um modelo com **alta variância** tende a **superajustar** (overfitting) os dados.  

* **Exemplo**: Um modelo complexo que prevê **exatamente os valores da amostra de treino**, mas erra muito nos novos dados.

A relação entre viés e variância é chamada de **trade-off viés-variância**.
